## 1. Chuỗi (String) _( Strings )_


Chuỗi là **immutable** (không đổi được). Mỗi thao tác 'sửa' chuỗi thực ra tạo ra một chuỗi mới.

### 1.1. Làm thế nào để đảo ngược chuỗi?

Có **3 cách** thông dụng. Pythonic nhất là **slicing với bước âm**.

- **Cách 1 — slicing `[::-1]`** (nhanh nhất, gọn nhất).
- **Cách 2 — `''.join(reversed(s))`** (ý đồ rõ ràng hơn).
- **Cách 3 — vòng lặp thủ công** (hiểu cách hoạt động bên trong).

![Slicing visualization](../assets/images/appendix/slicing.png)

---

### Phân tích cú pháp `[::-1]` chi tiết

Slicing tổng quát: `s[start:stop:step]` — **3 phần**, ngăn cách bởi dấu `:`. Cả 3 đều có thể bỏ qua, nếu bỏ thì Python dùng giá trị mặc định.

| Cú pháp | start | stop | step | Ý nghĩa |
|---------|-------|------|------|---------|
| `s[::]` | 0 | `len(s)` | 1 | toàn bộ chuỗi (tương đương `s`) |
| `s[::1]` | 0 | `len(s)` | 1 | giống trên |
| `s[::-1]` | `len(s)-1` | -∞ | -1 | đi ngược, từ cuối về đầu |
| `s[::2]` | 0 | `len(s)` | 2 | mỗi 2 ký tự lấy 1 (index 0, 2, 4...) |
| `s[1::2]` | 1 | `len(s)` | 2 | mỗi 2 ký tự lấy 1 (index 1, 3, 5...) |
| `s[::-2]` | `len(s)-1` | -∞ | -2 | đi ngược, mỗi 2 ký tự lấy 1 |

### Tại sao cần **2 dấu** `::`?

Trong slicing, dấu `:` ngăn cách các phần. Có 2 dạng:
- `s[a:b]` — chỉ định **start** và **stop** (chỉ 1 dấu `:`).
- `s[a:b:c]` — đầy đủ **start**, **stop**, **step** (2 dấu `:`).

`[::-1]` là dạng đầy đủ với `start` và `stop` bị **bỏ trống** (Python tự điền), chỉ rõ `step = -1`. Nếu chỉ viết `[:-1]` (1 dấu `:`) thì có nghĩa khác hoàn toàn — đó là `start=0, stop=-1` → bỏ ký tự cuối, không đảo!

### Tại sao `step = -1` đảo được chuỗi?

Khi `step` âm, Python **tự động** đặt `start = len(s) - 1` và `stop` ngầm trước index 0. Tức là duyệt từ ký tự cuối về đầu, mỗi bước -1 (lùi 1 ký tự).

### Thay số khác — kết quả ra sao?

Hãy chạy code phía dưới để kiểm chứng. Một số quy luật:
- **`step` dương**: đi từ trái sang phải, chỉ giữ phần tử cách nhau `step`.
- **`step` âm**: đi từ phải sang trái, tương tự.
- **`step = 0`**: lỗi `ValueError: slice step cannot be zero`.

### Vì sao slicing trả về **chuỗi mới**?

Chuỗi trong Python là **immutable** — không thể sửa tại chỗ. Mọi thao tác như đảo, cắt, thay thế đều **tạo chuỗi mới** trong RAM, không đụng đến chuỗi gốc.

### So sánh hiệu năng / Performance

- `s[::-1]` ⇒ **nhanh nhất** — slicing là thao tác C-level trong cài đặt CPython.
- `''.join(reversed(s))` ⇒ ~ngang bằng (`reversed()` lazy, `join` cấp phát 1 lần).
- Vòng lặp `kq = c + kq` ⇒ **chậm nhất** — O(n²), mỗi bước cấp phát chuỗi mới.


````python
s = 'Python'

# === 3 cách đảo chuỗi ===
print(s[::-1])                  # Cách 1: slicing
print(''.join(reversed(s)))     # Cách 2: reversed + join

kq = ''                         # Cách 3: vòng lặp thủ công
for c in s:
    kq = c + kq
print(kq)

# === Khám phá: thay số trong slicing ===
print(s[::1])     # 'Python'  — bước 1, không đảo
print(s[::-1])    # 'nohtyP'  — bước -1, đảo
print(s[::2])     # 'Pto'     — index 0, 2, 4
print(s[::-2])    # 'nhy'     — đi ngược, index 5, 3, 1
print(s[1::2])    # 'yhn'     — bắt đầu từ 1, mỗi 2 lấy 1
print(s[::-3])    # 'no'      — đi ngược, mỗi 3 lấy 1
print(s[2:5])     # 'tho'     — slice không có step
print(s[:-1])     # 'Pytho'   — bỏ ký tự cuối (KHÔNG đảo!)

# === Bẫy: step = 0 ===
# print(s[::0])   # ValueError: slice step cannot be zero
````

> Cách 1 là **chuẩn Python**. Hãy dùng nó trừ khi đang phỏng vấn cần thuật toán thủ công.

### 1.2. Làm sao kiểm tra một chuỗi có phải palindrome (đối xứng)?

Palindrome là chuỗi đọc xuôi và ngược đều giống nhau (`'aba'`, `'racecar'`).

Cách đơn giản nhất: so sánh `s == s[::-1]`. Nâng cao: bỏ qua dấu cách, chữ hoa/thường, dấu câu.

---

### Vì sao `s == s[::-1]` lại đúng?

- `s[::-1]` tạo ra **chuỗi mới đảo ngược** (xem câu 1.1).
- `==` so sánh **từng ký tự** của 2 chuỗi.
- Nếu chuỗi đối xứng, đọc xuôi và đọc ngược cho cùng dãy ký tự ⇒ `==` trả `True`.

Cách này gọn (1 dòng) nhưng dùng O(n) bộ nhớ phụ vì tạo chuỗi mới.

### Cách khác: two pointers (không tạo chuỗi mới)

![Two pointers cho palindrome](../assets/images/appendix/two_pointers.png)

Đặt 2 con trỏ ở 2 đầu, di chuyển vào giữa, so sánh từng cặp:
```python
def palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True
```
Cùng O(n) thời gian nhưng O(1) bộ nhớ — dừng ngay khi gặp khác nhau (không phải đợi tạo xong chuỗi đảo).

### Vì sao cần chuẩn hoá trước khi so sánh?

`'A man a plan a canal Panama'` — đọc bình thường thì là palindrome (theo nghĩa con người), nhưng `s == s[::-1]` trả `False` vì:
- Có dấu cách.
- Có chữ HOA / thường khác nhau.

Phải **chuẩn hoá**: bỏ ký tự không phải chữ/số (`isalnum()`), đổi về thường (`lower()`).

````python
def la_palindrome(s):
    return s == s[::-1]

print(la_palindrome('aba'))      # True
print(la_palindrome('python'))   # False

# Bỏ qua hoa thường, dấu cách:
def la_palindrome_chuan(s):
    chuan = ''.join(c for c in s.lower() if c.isalnum())
    return chuan == chuan[::-1]

print(la_palindrome_chuan('A man a plan a canal Panama'))  # True
````

### 1.3. Đếm số ký tự / số từ / số dòng?

- **Số ký tự**: `len(s)` (kể cả khoảng trắng và xuống dòng).
- **Số từ**: `len(s.split())`.
- **Số dòng**: `len(s.splitlines())` (an toàn với `\n` và `\r\n`).

---

### Vì sao `s.split()` (không có argument) lại an toàn?

`s.split()` (không truyền gì) khác `s.split(' ')` (truyền dấu cách):

| Cú pháp | Khi gặp | Kết quả với `'a  b   c'` |
|---------|---------|---------------------------|
| `s.split()` | gộp **mọi loại whitespace liên tiếp** (space, tab, newline) thành 1 sep | `['a', 'b', 'c']` |
| `s.split(' ')` | tách **từng** dấu cách → tạo string rỗng giữa các sep liên tiếp | `['a', '', 'b', '', '', 'c']` |

Đếm từ trong văn bản thực tế (có thể có nhiều khoảng trắng) → luôn dùng `split()` không tham số.

### Vì sao `splitlines()` lại tốt hơn `split('\n')`?

`splitlines()` xử lý **mọi cách xuống dòng** trên các hệ điều hành:
- `\n` — Unix / Linux / macOS
- `\r\n` — Windows
- `\r` — Mac cổ
- Còn cả `\v`, `\f`, `\x1c`, ... (hiếm)

Khác biệt cụ thể:
```python
s = 'A\nB\r\nC\nD\n'
print(s.split('\n'))      # ['A', 'B\r', 'C', 'D', '']  — có \r dư + '' cuối
print(s.splitlines())     # ['A', 'B', 'C', 'D']           — sạch sẽ
```

### Bẫy thường gặp

- `len(s.split('\n'))` cộng thêm 1 nếu file kết thúc bằng `\n` (vì có `''` ở cuối). `splitlines()` thì không.
- `len(s)` đếm cả `\n` — nếu chỉ muốn ký tự 'thấy được', lọc bằng `c.isprintable()`.

````python
s = 'Tôi yêu Python.\nHọc mỗi ngày.'
print('Ký tự:', len(s))
print('Từ:', len(s.split()))
print('Dòng:', len(s.splitlines()))
````

### 1.4. Đảo ngược thứ tự các từ (giữ nguyên ký tự trong từ)?

Chia chuỗi thành list các từ → đảo list → nối lại.
`'Tôi học Python'` → `'Python học Tôi'`.

---

### Phân tích từng bước

```python
s = 'Tôi học Python mỗi ngày'
tu = s.split()                # ['Tôi', 'học', 'Python', 'mỗi', 'ngày']
tu_dao = tu[::-1]             # ['ngày', 'mỗi', 'Python', 'học', 'Tôi']
kq = ' '.join(tu_dao)         # 'ngày mỗi Python học Tôi'
```
Gộp 1 dòng: `' '.join(s.split()[::-1])`.

### Phân biệt với `s[::-1]` (đảo ký tự)

Đừng nhầm:
- `s[::-1]` → `'yàgn iỗm nohtyP cọh iôT'` — đảo **từng ký tự**, không có nghĩa.
- `' '.join(s.split()[::-1])` → `'ngày mỗi Python học Tôi'` — đảo **các từ**, vẫn đọc được.

### Vì sao dùng `' '.join(...)` mà không phải `+`?

Cả 2 đều nối chuỗi nhưng `join` **nhanh và sạch hơn**:
- `+`: nối 2 chuỗi mỗi lần, mỗi lần cấp phát chuỗi mới ⇒ O(n²) với n từ.
- `join`: cấp phát **1 lần** đủ chỗ cho cả kết quả ⇒ O(n).
- `join` cũng đỡ phải xử lý dấu cách thừa ở 2 đầu (`'a' + ' ' + 'b'` vs `' '.join(['a','b'])`).

### Mẹo: muốn dùng dấu khác để nối

Đổi `' '` thành dấu phân cách bạn muốn:
```python
', '.join(['táo', 'cam', 'chuối'])    # 'táo, cam, chuối'
' | '.join(...)                       # 'táo | cam | chuối'
''.join(...)                          # 'táocamchuối' (không có sep)
```


````python
s = 'Tôi học Python mỗi ngày'
dao = ' '.join(s.split()[::-1])
print(dao)
````

### 1.5. Tách một chuỗi theo nhiều dấu phân cách (vd: dấu cách, phẩy, chấm phẩy)?

`str.split(sep)` chỉ tách theo **một** dấu. Để tách nhiều dấu, dùng module `re` (regular expression) với `re.split(pattern, s)`.

---

### Phân tích pattern `[,; ]+`

Đây là **regex** — ngôn ngữ mô tả mẫu chuỗi. Phân tích từng phần:

| Ký hiệu | Ý nghĩa |
|---------|---------|
| `[...]` | **Tập ký tự** (character class) — match **bất kỳ ký tự nào** trong ngoặc |
| `[,; ]` | match **dấu phẩy `,`**, hoặc **chấm phẩy `;`**, hoặc **dấu cách ` `** |
| `+` | **Một hoặc nhiều** lần lặp của ký tự ngay trước nó |
| `[,; ]+` | **Một hoặc nhiều** ký tự bất kỳ trong tập `, ; ` (gộp các sep liên tiếp) |

Vì sao cần `+`? Nếu input là `'táo,, cam'` (có 2 dấu phẩy):
- Pattern `[,; ]` (không `+`) → tách thành `['táo', '', ' cam']` (có '' giữa).
- Pattern `[,; ]+` (có `+`) → tách thành `['táo', 'cam']` (gộp các sep liên tiếp).

### Một số pattern hữu ích khác

| Pattern | Match gì |
|---------|----------|
| `\s+` | **Whitespace** (space, tab, newline) — gộp lại |
| `\d+` | **Một hoặc nhiều chữ số** |
| `\w+` | **Một hoặc nhiều** ký tự chữ/số/`_` |
| `[^,]+` | **Bất kỳ ký tự nào KHÁC dấu phẩy** (`^` ở đầu = phủ định) |

### Khi không cần regex

Nếu chỉ tách theo whitespace bất kỳ → dùng `s.split()` không có argument là đủ (xem câu 1.3). Chỉ cần regex khi có nhiều loại sep hoặc pattern phức tạp.

### Bẫy thường gặp

Một số ký tự là **đặc biệt** trong regex và cần escape bằng `\`:
`. ^ $ * + ? ( ) [ ] { } | \`
Ví dụ tách theo dấu chấm: `re.split(r'\.+', s)`.
Còn dùng prefix `r'...'` (raw string) để Python không nuốt mất dấu `\`.

````python
import re
s = 'táo, cam; chuối nhãn,xoài'
tu = re.split(r'[,; ]+', s)
print(tu)
````

### 1.6. So sánh 2 chuỗi không phân biệt hoa/thường (và bỏ khoảng trắng)?

Chuẩn hoá cả 2: `.strip().lower()` rồi mới so sánh.  
Để bỏ luôn dấu tiếng Việt, dùng `unicodedata.normalize`.

---

### Vì sao thứ tự `.strip().lower()` quan trọng?

Cả 2 thứ tự đều ra cùng kết quả vì:
- `.strip()` chỉ bỏ whitespace 2 đầu (không đụng vào ký tự chữ).
- `.lower()` chỉ đổi chữ HOA → thường (không đụng vào dấu cách).

Hai thao tác **không tương tác** với nhau — bạn có thể đổi thứ tự. Quy ước thường dùng `.strip().lower()` cho đọc tự nhiên (loại bỏ rác trước, biến đổi sau).

### Bỏ dấu tiếng Việt — `NFD` là gì?

Unicode có **2 cách biểu diễn** ký tự có dấu:
- **NFC (Composed)**: 1 ký tự duy nhất, vd `ế` lưu là 1 code point `U+1EBF`.
- **NFD (Decomposed)**: tách thành ký tự gốc + dấu, vd `ế` → `e` + ` ́` + ` ̂` (3 code points).

**Tactic của hàm `bo_dau`:**
1. `unicodedata.normalize('NFD', s)` → tách dấu ra khỏi ký tự gốc.
2. Lọc bỏ các 'dấu' (`category(c) == 'Mn'` — Mark Nonspacing).
3. Ghép lại bằng `''.join(...)` → còn lại chuỗi không dấu.

### Phân loại `category()` của Unicode

Mỗi ký tự Unicode có **danh mục 2 chữ cái**:

| Cat | Ý nghĩa | Ví dụ |
|-----|---------|-------|
| `Lu` | Letter, uppercase | `A`, `Б` |
| `Ll` | Letter, lowercase | `a`, `п` |
| `Mn` | **Mark, Nonspacing** | dấu sắc, dấu huyền (combining) |
| `Nd` | Number, decimal digit | `0`-`9` |
| `Pc` | Punctuation, connector | `_` |
| `Zs` | Separator, space | ` ` |

Ta lọc `Mn` vì sau khi `NFD`, các dấu thanh / dấu mũ tiếng Việt thuộc loại này.

### Ưu điểm vs nhược điểm

**Ưu**: 1 hàm xử lý mọi loại dấu (Việt, Pháp, Đức, ...) — generic.
**Nhược**: chậm hơn dùng `str.translate()` với bảng ánh xạ tay; không xử lý đúng `đ` → `d` (vì `đ` không phân tách được trong NFD).

````python
a = '  Xin Chào  '
b = 'xin chào'
print(a.strip().lower() == b.strip().lower())   # True

import unicodedata
def bo_dau(s):
    nfd = unicodedata.normalize('NFD', s)
    return ''.join(c for c in nfd if unicodedata.category(c) != 'Mn')

print(bo_dau('Xin chào tiếng Việt'))   # 'Xin chao tieng Viet'
````

### 1.7. Định dạng số đẹp trong f-string (làm tròn, đệm số 0, %, mũ)?

Cú pháp: `f'{value:format_spec}'`.

| Spec | Ý nghĩa |
|------|---------|
| `:.2f` | float, 2 chữ số thập phân |
| `:,.2f` | có dấu phẩy phân cách hàng nghìn |
| `:5d` | int, đệm khoảng trắng đến rộng 5 |
| `:05d` | int, đệm 0 đến rộng 5 |
| `:.1%` | tỉ lệ (nhân 100) |
| `:e` | dạng khoa học |
| `:>10` / `:<10` / `:^10` | căn phải / trái / giữa, rộng 10 |

---

### Ngữ pháp tổng quát của format spec

Sau dấu `:` trong f-string là **mini-language** với cấu trúc:

```
[fill][align][sign][#][0][width][grouping][.precision][type]
```

Mỗi ô đều **tuỳ chọn**. Đọc từ trái sang phải:

| Phần | Ý nghĩa | Ví dụ |
|------|---------|-------|
| `fill` | ký tự lấp đầy (1 ký tự) | `*` trong `:*>10` |
| `align` | căn lề: `<` trái, `>` phải, `^` giữa, `=` (cho số) | `:>10` |
| `sign` | dấu: `+` luôn hiện, `-` chỉ hiện cho âm, ` ` (space) — dấu cách trước số dương | `:+.2f` |
| `#` | dạng 'alternate' (cho hex/oct/bin: thêm `0x`, `0o`, `0b`) | `:#x` |
| `0` | đệm số 0 thay vì khoảng trắng | `:05d` |
| `width` | tổng độ rộng tối thiểu | `:10d` |
| `grouping` | `,` phân cách hàng nghìn, `_` cũng được | `:,.2f` |
| `.precision` | số chữ số sau dấu chấm (float) hoặc số ký tự (str) | `:.3f` |
| `type` | kiểu hiển thị: `f`, `d`, `e`, `%`, `b`, `o`, `x`, `s`... | `:.2f` |

### Phân tích từng ví dụ

- `:.2f` ⇒ precision=2, type=`f` (float). Vd `3.14159` → `'3.14'`.
- `:,.2f` ⇒ grouping=`,`, precision=2, type=`f`. Vd `12345.6` → `'12,345.60'`.
- `:05d` ⇒ width=5, đệm số 0, type=`d` (decimal int). Vd `42` → `'00042'`.
- `:.1%` ⇒ precision=1, type=`%` (nhân 100, thêm dấu %). Vd `0.873` → `'87.3%'`.
- `:e` ⇒ type=`e` (scientific). Vd `12345` → `'1.234500e+04'`.
- `:>10` ⇒ align=`>`, width=10. Vd `'hi'` → `'        hi'`.

### Các kiểu (`type`) hữu ích

| Type | Áp dụng cho | Ví dụ |
|------|-------------|-------|
| `d` | int dạng thập phân | `42` |
| `b` | int dạng binary | `101010` |
| `o` | int dạng octal | `52` |
| `x` / `X` | int dạng hex (chữ thường / HOA) | `2a` / `2A` |
| `f` | float dạng thập phân | `3.14` |
| `e` / `E` | float dạng khoa học | `1.23e+02` / `1.23E+02` |
| `g` | float, tự chọn `f` hay `e` cho gọn | `1.23` hoặc `1.23e+10` |
| `%` | float × 100 + dấu `%` | `87.3%` |
| `s` | string (mặc định) | `hello` |

### Mẹo nhỏ thường dùng

- `f'{x=}'` (Python 3.8+) — tự động in `'x = giá_trị'` (debug nhanh).
- `f'{x:.{n}f}'` — precision **động** (lấy từ biến `n`).
- `f'{value!r}'` — gọi `repr(value)` thay vì `str(value)` (hữu ích để in chuỗi có dấu nháy).

````python
x = 12345.6789
print(f'{x:.2f}')        # 12345.68
print(f'{x:,.2f}')       # 12,345.68
print(f'{42:05d}')       # 00042
print(f'{0.873:.1%}')    # 87.3%
print(f'{x:.2e}')        # 1.23e+04
print(f'{"Tên":<10}{"Điểm":>8}')
print(f'{"An":<10}{8.5:>8.1f}')
print(f'{"Bình":<10}{9.0:>8.1f}')
````

### 1.8. Mã hoá Caesar đơn giản (dịch ký tự đi N vị trí)?

Dùng `ord()` để lấy mã ASCII và `chr()` để chuyển ngược lại. Phép `% 26` để 'cuốn vòng' khi vượt 'z'.

---

### Hàm `ord()` và `chr()` là gì?

Mỗi ký tự máy tính có **mã số** (code point):
- `ord(c)` → mã số (int) của ký tự `c`. `ord('A')` = 65, `ord('a')` = 97.
- `chr(n)` → ký tự (str) của mã số `n`. `chr(65)` = `'A'`.

Bảng ASCII:
- `'A'` đến `'Z'` → 65 đến 90 (liền nhau).
- `'a'` đến `'z'` → 97 đến 122 (liền nhau).

### Phân tích công thức cho chữ HOA

```python
chr((ord(c) - ord('A') + n) % 26 + ord('A'))
```

Đọc từ trong ra ngoài:
1. `ord(c) - ord('A')` ⇒ đổi từ ASCII (65-90) về **index 0-25** (A=0, B=1, ..., Z=25).
2. `+ n` ⇒ dịch đi `n` vị trí (có thể vượt 25 hoặc âm).
3. `% 26` ⇒ **cuốn vòng** trong khoảng [0, 26). Vd `26 % 26 = 0`, `27 % 26 = 1`, ..., `-1 % 26 = 25`.
4. `+ ord('A')` ⇒ đổi index 0-25 trở lại ASCII 65-90.
5. `chr(...)` ⇒ ASCII trở lại ký tự.

### Vì sao `% 26` lại 'cuốn vòng' an toàn cả với `n` âm?

Trong Python, **modulo của số âm cũng dương** (khác C/Java):
- `-1 % 26` → `25` (Python).
- `-1 % 26` → `-1` (C, Java).

Đây là một trong những điểm Python 'thân thiện' hơn — `caesar('A', -1)` ra `'Z'` ngay, không cần xử lý đặc biệt cho `n` âm.

### Vì sao xử lý hoa và thường **riêng biệt**?

ASCII của `'A'` (65) và `'a'` (97) cách nhau 32. Nếu cộng tất cả vào cùng một dải, khi cuốn vòng từ `'Z'` sẽ rơi xuống `'A'` đúng → OK. Nhưng `'z'` đi tiếp 1 bước sẽ rơi vào ký tự lạ (`'{'` = 123), không phải `'a'`. Phải tách hoa thường ra để mỗi dải tự cuốn vòng đúng phạm vi của nó.

### Bẫy & mở rộng

- **Tiếng Việt**: hàm này không xử lý đúng cho `'á'`, `'đ'`... vì chúng nằm ngoài dải ASCII Latin. Cần map riêng hoặc bỏ dấu trước khi mã hoá.
- **Bảo mật**: Caesar **cực kỳ yếu** (chỉ 25 khả năng — brute force trong 1 giây). Đừng dùng cho mật khẩu thật.

````python
def caesar(s, n):
    kq = []
    for c in s:
        if c.isupper():
            kq.append(chr((ord(c) - ord('A') + n) % 26 + ord('A')))
        elif c.islower():
            kq.append(chr((ord(c) - ord('a') + n) % 26 + ord('a')))
        else:
            kq.append(c)
    return ''.join(kq)

print(caesar('Hello', 3))         # Khoor
print(caesar('Khoor', -3))        # Hello
````

## 2. List và Tuple _( Lists and Tuples )_


**List** mutable, dùng `[ ]`. **Tuple** immutable, dùng `( )`. Dưới đây là các thao tác thường gặp.

### 2.1. Tìm min / max / sum / trung bình của list?

Dùng các hàm built-in. Trung bình tự tính bằng `sum / len`.

---

### Vì sao Python không có `avg()` built-in?

`min`, `max`, `sum` đều **không phụ thuộc vào kiểu dữ liệu** — chỉ cần biết `<` và `+`. Còn 'trung bình' lại có nhiều biến thể: trung bình cộng, trung bình hình học, trung bình điều hoà, trung vị... → Python tách ra module `statistics` thay vì gắn 1 hàm `avg` cố định.

Cách an toàn:
```python

```

```python
min(tu, key=len)        # 'ổi' — từ ngắn nhất
```


````python
ds = [3, 7, 1, 9, 5, 2]
print('Min:', min(ds))
print('Max:', max(ds))
print('Sum:', sum(ds))
print('Trung bình:', sum(ds) / len(ds))
````

### 2.2. Loại bỏ trùng lặp trong list?

**Cách nhanh**: `list(set(ds))` — nhưng **mất thứ tự**.  
**Giữ thứ tự**: dùng `dict.fromkeys()` (từ Python 3.7 dict giữ thứ tự chèn) hoặc vòng lặp + `seen` set.

---

### Vì sao `set` mất thứ tự?

`set` được cài đặt bằng **hash table** — vị trí của phần tử trong bộ nhớ phụ thuộc vào **hash của giá trị**, không phải thứ tự bạn chèn vào. Đây là lý do `in` trên set là O(1) — nhưng cũng là lý do `for x in set(...)` có thể duyệt theo thứ tự khác mỗi lần chạy.

### Vì sao `dict.fromkeys()` lại giữ thứ tự?

Từ Python 3.7+, `dict` **bảo đảm giữ thứ tự chèn** (đây là spec của ngôn ngữ, không chỉ chi tiết cài đặt). `dict.fromkeys(ds)` tạo dict với:
- key = từng phần tử của `ds` (lần đầu xuất hiện sẽ được giữ).
- value = `None` (không quan tâm).

Khi convert lại thành `list`, ta lấy **các key** theo thứ tự chèn → bằng thứ tự xuất hiện đầu tiên trong `ds`.

### Cách thủ công với `seen` set — nguyên lý

```python
seen = set()      # chỉ để check 'đã thấy chưa' — O(1)
moi = []          # list kết quả — giữ thứ tự
for x in ds:
    if x not in seen:    # O(1) thay vì O(n) nếu dùng `if x not in moi`
        seen.add(x)
        moi.append(x)
```

Kết hợp ưu điểm của 2 cấu trúc: set cho check nhanh, list cho thứ tự.

### So sánh độ phức tạp

| Cách | Thời gian | Bộ nhớ | Giữ thứ tự |
|------|-----------|--------|------------|
| `list(set(ds))` | O(n) | O(n) | ❌ |
| `list(dict.fromkeys(ds))` | O(n) | O(n) | ✓ |
| Vòng lặp + seen | O(n) | O(n) | ✓ |
| `[x for i, x in enumerate(ds) if x not in ds[:i]]` | O(n²) | O(1) phụ | ✓ — **chậm** |

Cách cuối **không nên dùng** — chậm vì mỗi bước check `not in ds[:i]` mất O(i).

### Bẫy: phần tử không hashable

`set` và `dict.fromkeys` chỉ hoạt động với phần tử **hashable** (int, str, tuple). Nếu list chứa list/dict bên trong:
```python
ds = [[1, 2], [3], [1, 2]]
set(ds)  # TypeError: unhashable type: 'list'
```
Phải dùng cách thủ công với check `if x not in moi` (O(n²)) hoặc convert thành tuple trước.

````python
ds = [3, 1, 3, 2, 1, 4, 2, 5]

# Mất thứ tự:
print(list(set(ds)))

# Giữ thứ tự — Pythonic:
print(list(dict.fromkeys(ds)))

# Giữ thứ tự — thủ công:
seen = set()
moi = []
for x in ds:
    if x not in seen:
        seen.add(x)
        moi.append(x)
print(moi)
````

### 2.3. Sắp xếp list theo một tiêu chí tuỳ ý (tên, độ dài, trường của dict)?

Dùng tham số `key=` của `sorted()` hoặc `list.sort()`. `key` là một hàm trả về 'điểm' để so sánh — thường là `lambda`.

---

### `key=` hoạt động thế nào?

`key=fn` nói với `sorted` rằng: "khi cần so sánh 2 phần tử, hãy gọi `fn(x)` để lấy điểm, rồi so sánh các **điểm** chứ không phải phần tử gốc".

```python
tu = ['cherry', 'apple', 'banana']
sorted(tu, key=len)   # so sánh len('cherry')=6 vs len('apple')=5 vs len('banana')=6
                      # ⇒ ['apple', 'cherry', 'banana']
```

Ưu điểm: hàm `fn` chỉ chạy **1 lần cho mỗi phần tử** (Python cache lại) — gọi là *Schwartzian transform* / decorate-sort-undecorate.

### `sort()` vs `sorted()` — khác gì?

| | `list.sort()` | `sorted(iterable)` |
|--|-------------|---------------------|
| Sửa list gốc? | ✓ có | ❌ không |
| Trả về | `None` | list mới |
| Dùng cho | list | bất kỳ iterable (set, dict, generator...) |

**Lưu ý:** `ds = ds.sort()` là **lỗi phổ biến** — vì `sort()` trả `None`, gán xong `ds = None`!

### `reverse=True` — đảo thứ tự sau khi sort

`sorted(tu, key=len, reverse=True)` ⇒ sort tăng theo `key`, sau đó đảo → giảm dần.  
Tương đương với `sorted(tu, key=lambda x: -len(x))` cho số, nhưng `reverse=True` tổng quát hơn (chạy với mọi kiểu, kể cả không có phép `-`).

### Sort theo **nhiều tiêu chí**

Trả về **tuple** từ `key` — Python so sánh tuple theo thứ tự từ trái sang phải:

```python
hs = [
    {'ten': 'An',   'lop': '10A1', 'diem': 8.5},
    {'ten': 'Bình', 'lop': '10A1', 'diem': 7.0},
    {'ten': 'Châu', 'lop': '10A2', 'diem': 8.5},
]
# Sort: theo lớp tăng dần, trong cùng lớp theo điểm GIẢM dần
sorted(hs, key=lambda h: (h['lop'], -h['diem']))
```

### Stable sort — đặc tính quan trọng

Sort của Python là **stable**: nếu 2 phần tử có cùng `key`, thứ tự gốc của chúng được giữ. Vd: sort 1 list theo `lop`, các bạn cùng lớp vẫn theo thứ tự xuất hiện ban đầu.

Áp dụng: muốn sort theo nhiều tiêu chí, có thể **sort nhiều lần từ ít quan trọng đến quan trọng nhất**:
```python
# Sort theo điểm giảm dần TRƯỚC, rồi theo lớp tăng dần SAU
hs.sort(key=lambda h: -h['diem'])
hs.sort(key=lambda h: h['lop'])
```


````python
tu = ['cherry', 'apple', 'banana', 'fig']

print(sorted(tu))                      # alphabet
print(sorted(tu, key=len))             # theo độ dài
print(sorted(tu, key=len, reverse=True))

hs = [
    {'ten': 'An', 'diem': 8.5},
    {'ten': 'Bình', 'diem': 7.0},
    {'ten': 'Châu', 'diem': 9.0},
]
by_diem = sorted(hs, key=lambda h: h['diem'], reverse=True)
for h in by_diem:
    print(h['ten'], h['diem'])
````

### 2.4. Làm phẳng một list 2D (flatten)?

List comprehension lồng là Pythonic nhất. Cũng có `itertools.chain`.

---

### Đọc comprehension lồng từ trái sang phải

```python
[x for hang in M for x in hang]
```

Đọc theo thứ tự y hệt vòng `for` thường:
```python
for hang in M:        # vòng ngoài: lặp qua từng hàng
    for x in hang:    # vòng trong: lặp qua từng phần tử của hàng
        # x là phần tử cần thêm vào list mới
```

Quy tắc: **biến chạy đầu tiên (`for hang in M`) là vòng ngoài**, biến tiếp theo là vòng trong. **Biểu thức sinh ra giá trị (`x`) đặt ở đầu**.

### Vì sao không phải `[x for x in hang for hang in M]`?

Đảo thứ tự sẽ gây **NameError** — biến `hang` chưa được định nghĩa khi vòng trong chạy. Quy tắc nhớ nhanh: viết comprehension **giống vòng for thường** (ngoài → trong → biểu thức), rồi *kéo biểu thức lên đầu*.

### So sánh các cách flatten

| Cách | Code | Khi nào nên dùng |
|------|------|-------------------|
| Comprehension | `[x for r in M for x in r]` | Pythonic, ngắn, ai cũng hiểu |
| `chain.from_iterable` | `list(chain.from_iterable(M))` | Khi M lớn (lazy) hoặc không phải list |
| `sum` | `sum(M, [])` | **Tránh** — O(n²), chậm khủng khiếp |
| numpy | `np.array(M).flatten()` | Đã dùng numpy, ma trận đều hàng |

### Bẫy: chỉ flatten 1 cấp

Comprehension `[x for r in M for x in r]` chỉ làm phẳng **1 cấp**. Với list 3D như `[[[1,2],[3]],[[4]]]`, nó cho ra `[[1,2],[3],[4]]` (vẫn còn 1 cấp). Để flatten sâu (recursive), cần viết hàm đệ quy:
```python
def flatten(x):
    if isinstance(x, list):
        return [a for item in x for a in flatten(item)]
    return [x]
```


````python
M = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]

phang = [x for hang in M for x in hang]
print(phang)

# Hoặc:
from itertools import chain
print(list(chain.from_iterable(M)))
````

### 2.5. Chia list thành các nhóm n phần tử (chunk)?

Slicing trong vòng lặp với bước n.  
Hữu ích để chia đề / chia nhóm đều.

---

### Phân tích `range(0, len(ds), n)`

`range(start, stop, step)` sinh dãy `start, start+step, start+2*step, ...` (dừng trước `stop`).

Vd với `len(ds) = 10, n = 3`:
- `range(0, 10, 3)` → `0, 3, 6, 9`
- `ds[0:3]`, `ds[3:6]`, `ds[6:9]`, `ds[9:12]` (slice tự xử lý vượt biên thành slice ngắn hơn)
- Kết quả: 4 chunks — 3 chunks 3 phần tử + 1 chunk 1 phần tử cuối.

### Vì sao slicing **không** raise lỗi khi vượt biên?

Khác với `ds[i]` (raise `IndexError` nếu `i >= len`), slicing `ds[a:b]` **tự cắt** về độ dài thực tế. Đây là tính chất 'tha thứ' của slicing — rất tiện cho chunk vì không cần check biên.

### Chunk lazy với generator

Nếu data lớn (vd 1 triệu phần tử), tạo cả list chunks tốn RAM. Generator lazy hơn:
```python
def chunks(ds, n):
    for i in range(0, len(ds), n):
        yield ds[i:i + n]

for c in chunks(big_list, 1000):
    process(c)   # chỉ giữ 1 chunk trong RAM tại một thời điểm
```


````python
def chunks(ds, n):
    return [ds[i:i + n] for i in range(0, len(ds), n)]

ds = list(range(1, 11))
print(chunks(ds, 3))
````

### 2.6. Tìm phần tử lớn thứ 2?

**Cách 1**: `sorted(set(ds))[-2]` — gọn, nhưng O(n log n).  
**Cách 2**: duy trì 2 biến `lon1, lon2` — O(n).

---

### Vì sao cần `set()` trong cách 1?

Nếu list có giá trị **trùng** (vd `[9, 9, 8, 7]`), `sorted(ds)[-2]` cho kết quả `9` (vì có 2 số 9) — không phải 'lớn thứ 2 *khác nhau*'. `set` loại trùng trước khi sort, đảm bảo `[-2]` đúng nghĩa.

### Phân tích cách 2: O(n) một-vòng

Nguyên lý: duy trì **invariant** `lon1 ≥ lon2 ≥ tất_cả_đã_thấy`. Mỗi phần tử mới chỉ cần so với `lon1` và `lon2`:
```python
lon1 = lon2 = float('-inf')
for x in ds:
    if x > lon1:        # x lớn hơn cả max hiện tại
        lon2 = lon1     # max cũ trở thành 2nd
        lon1 = x        # x là max mới
    elif lon1 > x > lon2:   # x giữa lon2 và lon1
        lon2 = x        # cập nhật 2nd
```

Vì sao `float('-inf')` (số âm vô cùng)?  
Để mọi phần tử thật trong list **đều lớn hơn** giá trị khởi tạo. Nếu khởi tạo `lon1 = lon2 = 0`, list toàn số âm sẽ bug (xem câu 7.4 trong S07).

### Bẫy: `lon1 > x > lon2` ⇒ loại bỏ giá trị bằng `lon1`

Nếu list là `[9, 9, 8]`, cách trên trả về `8` (đúng nghĩa 'lớn thứ 2 khác nhau'). Nếu muốn **'lớn thứ 2 kể cả trùng'** (trả 9 cho list `[9, 9, 8]`), đổi điều kiện thành `elif x > lon2 and x != lon1`. Tuỳ ngữ cảnh muốn gì.

### Mở rộng: top-K phần tử

Cho K nhỏ, `heapq` cực kỳ hiệu quả — O(n log K) thay vì O(n log n) của sort:
```python
import heapq
top3 = heapq.nlargest(3, ds)
```


````python
ds = [3, 8, 1, 9, 4, 9, 7]

# Cách 1:
print(sorted(set(ds))[-2])

# Cách 2:
lon1 = lon2 = float('-inf')
for x in ds:
    if x > lon1:
        lon2 = lon1
        lon1 = x
    elif lon1 > x > lon2:
        lon2 = x
print(lon2)
````

### 2.7. Đếm số lần xuất hiện của mỗi phần tử?

Dùng `collections.Counter` — chuẩn nhất.  
Không cần import: dùng dict + `.get(x, 0) + 1`.

---

### `Counter` là gì?

`Counter` là **subclass của `dict`** chuyên cho việc đếm:
- Tự khởi tạo value = 0 khi truy cập key chưa có (không raise `KeyError`).
- Có sẵn method `most_common(k)` trả về top-K theo count, đã sort sẵn.
- Hỗ trợ phép `+`, `-`, `&` (giao), `|` (hợp) giữa 2 Counter.

### Vì sao `dem[c] = dem.get(c, 0) + 1` phải dùng `.get`?

`dem[c]` raise `KeyError` nếu `c` chưa có. `dem.get(c, 0)` trả về 0 cho key không có — an toàn.

Cách khác cùng ý: `defaultdict(int)`:
```python
from collections import defaultdict
dem = defaultdict(int)
for x in ds:
    dem[x] += 1   # int() = 0 → tự khởi tạo
```

### Phép toán giữa 2 `Counter`

Cực kỳ hữu ích cho 'so sánh anagram', 'tìm chữ cần thêm để biến A thành B', ...
```python
from collections import Counter
a = Counter('listen')
b = Counter('silent')
print(a == b)        # True — anagram

c = Counter('aabbcc')
d = Counter('abc')
print(c - d)         # Counter({'a': 1, 'b': 1, 'c': 1}) — phần c thừa so với d
```


````python
ds = ['táo', 'cam', 'táo', 'lê', 'cam', 'táo']

# Pythonic:
from collections import Counter
print(Counter(ds))
print(Counter(ds).most_common(2))   # 2 phần tử phổ biến nhất

# Không import:
dem = {}
for x in ds:
    dem[x] = dem.get(x, 0) + 1
print(dem)
````

### 2.8. Đảo ngược list (3 cách)?

- `ds[::-1]` — tạo list mới (Pythonic).
- `ds.reverse()` — sửa list gốc.
- `reversed(ds)` — trả về iterator (cần `list()` để in).

---

### So sánh chi tiết

| Cách | Trả về | Sửa gốc? | Bộ nhớ | Khi nào dùng |
|------|--------|----------|--------|---------------|
| `ds[::-1]` | list mới | ❌ | O(n) — copy toàn bộ | cần list mới, không đụng gốc |
| `ds.reverse()` | `None` | ✓ | O(1) — đảo tại chỗ | tiết kiệm RAM, không cần list cũ |
| `reversed(ds)` | iterator | ❌ | O(1) — lazy | chỉ cần duyệt 1 lần (vd `for x in reversed(ds)`) |

### Vì sao 3 cách tồn tại nếu cách 1 đã đủ?

- **Hiệu năng** với data lớn: `ds.reverse()` không cấp phát thêm bộ nhớ.
- **Lazy evaluation**: `reversed(ds)` không tạo list mới — nếu bạn chỉ duyệt qua, không cần copy.
- **Ý đồ rõ ràng**: `reversed(ds)` đọc 'đảo ngược' rõ hơn `[::-1]` cho người chưa quen Python.

### Bẫy: `ds.reverse()` trả `None`

```python
ds = [1, 2, 3]
moi = ds.reverse()    # ❌ moi là None
print(moi)            # None
print(ds)             # [3, 2, 1] — ds đã bị sửa
```

Quy tắc Python: các **method sửa list tại chỗ** (`sort`, `reverse`, `append`, `extend`, ...) đều trả `None` để **nhắc bạn** rằng nó sửa gốc. Nếu thấy `x = ds.sort()` → 99% là bug.

````python
ds = [1, 2, 3, 4, 5]
print(ds[::-1])              # [5,4,3,2,1] — list mới
print(list(reversed(ds)))    # tương tự
ds.reverse()
print(ds)                    # ds đã bị sửa
````

### 2.9. Khi nào dùng `list` vs `tuple`?

**Dùng `tuple` khi:**
- Dữ liệu **không nên đổi** (toạ độ, ngày, RGB color).
- Trả về nhiều giá trị từ hàm.
- Làm **key của dict** (list không làm được).

**Dùng `list` khi:**
- Cần thêm/bớt/sửa phần tử.
- Số lượng phần tử có thể thay đổi.

**Mẹo:** tuple thường nhanh hơn list một chút và tốn ít bộ nhớ hơn.

![Bố cục bộ nhớ list / tuple / dict](../assets/images/appendix/data_structures.png)

---

### Vì sao tuple dùng được làm key của dict mà list thì không?

Dict cần **hash** key để xác định vị trí lưu. Hash phải **không đổi theo thời gian** (nếu key đổi, dict không tìm lại được). 

- **Tuple immutable** ⇒ giá trị cố định ⇒ hash cố định ⇒ làm key OK.
- **List mutable** ⇒ `ds.append(x)` đổi nội dung ⇒ hash sẽ đổi ⇒ Python cấm làm key (chứ không phải technical limit).

### Ngữ nghĩa: tuple = 'record', list = 'collection'

Có một quy tắc đẹp:
- **Tuple** dùng cho dữ liệu mà **mỗi vị trí có ý nghĩa khác nhau**: `(x, y)` — vị trí 0 là x, vị trí 1 là y. **Heterogeneous**.
- **List** dùng cho dãy các phần tử **cùng loại**, có thể có 0, 1, hay 1000 phần tử. **Homogeneous**.

Vd: list các điểm trong không gian = `list of tuples`:
```python
diem = [(0, 0), (1, 2), (3, 4)]
```

### Tuple có hỗ trợ method gì?

Vì immutable, tuple chỉ có 2 method:
- `t.count(x)` — đếm số lần `x` xuất hiện.
- `t.index(x)` — vị trí đầu của `x`.

Còn lại (`append`, `pop`, `remove`, `sort`...) đều **không có** vì sẽ phải đổi tuple.

### Hiệu năng thực tế

Tuple ~5-15% nhanh hơn list trong các phép cơ bản (tạo, truy cập index, iteration). Tốn ít bộ nhớ hơn (~10-20 bytes/phần tử thay vì ~28). Khác biệt nhỏ — chỉ đáng cân nhắc với data triệu phần tử.

````python
# Tuple làm key của dict
vi_tri = {
    (0, 0): 'gốc',
    (1, 0): 'phải',
    (0, 1): 'trên',
}
print(vi_tri[(1, 0)])

# List KHÔNG làm key được:
# d = {[1, 0]: 'x'}   # TypeError
````

## 3. Dictionary và Set _( Dictionaries and Sets )_


**Dict** ánh xạ key → value, tra cứu O(1). **Set** lưu phần tử duy nhất, kiểm tra `in` cũng O(1). Cả hai đều **không có thứ tự** theo nghĩa logic (dù từ Python 3.7+ dict giữ thứ tự chèn).

### 3.1. Đếm tần suất các phần tử trong list/chuỗi?

Pythonic nhất là `Counter`. Nếu không muốn import, dùng `dict.get(x, 0) + 1`.

---

### Vì sao mẫu `dict.get(c, 0) + 1` lại đẹp?

Phân tích từng phần:
- `dict.get(c, 0)` — trả value của key `c`. Nếu chưa có, **trả 0** (không raise lỗi).
- `... + 1` — cộng 1 cho count.
- `dem[c] = ...` — gán lại (nếu chưa có thì tạo mới với giá trị 1).

Toàn bộ logic 'đã có thì +1, chưa có thì = 1' gói gọn trong 1 dòng.

### `most_common(k)` của `Counter`

Trả list các tuple `(phần_tử, count)` đã **sort theo count giảm dần**.

```python
Counter('mississippi').most_common(2)
# [('i', 4), ('s', 4)]
```

Cài đặt nội bộ dùng `heapq.nlargest` ⇒ O(n log k) thay vì O(n log n) của sort toàn bộ.

### Khi `Counter` không phù hợp

- Nếu chỉ cần **biết có/không** (không cần đếm) → `set` đủ.
- Nếu cần đếm có **trọng số** (vd phần tử a có weight 0.7) → tự dùng dict + `dem.get(c, 0) + weight`.


````python
from collections import Counter

s = 'mississippi'
print(Counter(s))
print(Counter(s).most_common(2))   # top 2 ký tự

# Không import:
dem = {}
for c in s:
    dem[c] = dem.get(c, 0) + 1
print(dem)
````

### 3.2. Group by — gom các phần tử theo một key?

Dùng `dict` với `setdefault(key, []).append(value)` hoặc `collections.defaultdict(list)`. Ví dụ: gom điểm theo lớp.

---

### `defaultdict(list)` hoạt động thế nào?

`defaultdict(factory)` là `dict` với trick: khi truy cập key chưa có, nó **tự động gọi `factory()`** để tạo value mặc định.

Với `factory=list`:
```python
from collections import defaultdict
d = defaultdict(list)
d['a'].append(1)   # 'a' chưa có → tự tạo d['a'] = [], rồi append → [1]
d['a'].append(2)   # 'a' đã có → append → [1, 2]
print(d)           # defaultdict(<class 'list'>, {'a': [1, 2]})
```

Khác `dict` thường: với `dict`, dòng `d['a'].append(1)` raise `KeyError`.

### So sánh 3 cách group

```python
# Cách 1 — defaultdict (Pythonic)
from collections import defaultdict
g = defaultdict(list)
for h in hs:
    g[h['lop']].append(h['ten'])

# Cách 2 — dict.setdefault
g = {}
for h in hs:
    g.setdefault(h['lop'], []).append(h['ten'])

# Cách 3 — kiểm tra thủ công
g = {}
for h in hs:
    if h['lop'] not in g:
        g[h['lop']] = []
    g[h['lop']].append(h['ten'])
```

Cả 3 ra cùng kết quả. Cách 1 đẹp nhất khi import được; cách 2 không cần import.

### Nâng cao: `itertools.groupby`

Khác `defaultdict`: yêu cầu **list đã sort theo key** trước. Hữu ích cho data đã sắp:
```python
from itertools import groupby
hs.sort(key=lambda h: h['lop'])
for lop, items in groupby(hs, key=lambda h: h['lop']):
    print(lop, list(items))
```
Lazy — chỉ giữ 1 group trong RAM tại một thời điểm.

````python
from collections import defaultdict

hs = [
    {'ten': 'An', 'lop': '10A1', 'diem': 8.5},
    {'ten': 'Bình', 'lop': '10A2', 'diem': 7.0},
    {'ten': 'Châu', 'lop': '10A1', 'diem': 9.0},
    {'ten': 'Dương', 'lop': '10A2', 'diem': 6.5},
]

theo_lop = defaultdict(list)
for h in hs:
    theo_lop[h['lop']].append(h['ten'])

for lop, ten in theo_lop.items():
    print(lop, '->', ten)

# Trung bình điểm theo lớp:
from statistics import mean
tb_lop = defaultdict(list)
for h in hs:
    tb_lop[h['lop']].append(h['diem'])
for lop, ds in tb_lop.items():
    print(lop, mean(ds))
````

### 3.3. Sắp xếp dict theo value?

Dict không 'sắp xếp' tại chỗ — phải tạo list các cặp `(key, value)` đã sort. Sau đó có thể chuyển ngược về dict (dict giữ thứ tự chèn).

---

### Phân tích `sorted(d.items(), key=lambda kv: kv[1])`

- `d.items()` ⇒ tạo view các tuple `(key, value)`. Vd `dict_items([('An', 8.5), ('Bình', 7.0)])`.
- `key=lambda kv: kv[1]` ⇒ với mỗi tuple `kv`, lấy phần tử thứ 1 (value) làm tiêu chí sort.
- `sorted(...)` ⇒ trả list mới đã sort. Mặc định tăng dần.

### Tại sao **không** thể `dict.sort()`?

Trước Python 3.7, dict **không có thứ tự** — sort không có ý nghĩa. Từ 3.7+, dict giữ thứ tự chèn nhưng vẫn không có method `.sort()` vì philosophy: **dict dành cho tra cứu O(1)**, sort là việc của list.

### Sort key thay vì value

```python
# Theo key (alphabet):
sorted(d.items())                          # mặc định sort theo phần tử đầu của tuple = key
sorted(d.items(), key=lambda kv: kv[0])    # tương đương

# Theo value (đã thấy):
sorted(d.items(), key=lambda kv: kv[1])

# Theo độ dài key:
sorted(d.items(), key=lambda kv: len(kv[0]))
```

### Dùng `operator.itemgetter` thay lambda

Hơi nhanh hơn lambda (do C-level):
```python
from operator import itemgetter
sorted(d.items(), key=itemgetter(1))   # tương đương lambda kv: kv[1]
```


````python
diem = {'An': 8.5, 'Bình': 7.0, 'Châu': 9.0, 'Dương': 6.5}

# Tăng dần theo value:
tang = sorted(diem.items(), key=lambda kv: kv[1])
print(tang)

# Giảm dần, chuyển về dict:
giam = dict(sorted(diem.items(), key=lambda kv: kv[1], reverse=True))
print(giam)
````

### 3.4. Gộp 2 dict (merge)?

Python 3.9+: dùng toán tử `|`. Trước đó: `{**d1, **d2}` hoặc `d1.update(d2)`.

Khi trùng key, **dict bên phải thắng**.

---

### Toán tử `|` cho dict (Python 3.9+)

Lấy ý tưởng từ set union — `a | b` cho dict tạo dict mới với:
- Tất cả key của `a`.
- Cộng thêm key của `b`.
- Key trùng: lấy giá trị từ `b` (bên phải).

Tương tự có `|=` để cập nhật tại chỗ:
```python
a = {'x': 1}
a |= {'y': 2}        # a giờ là {'x': 1, 'y': 2}
```

### `{**a, **b}` — unpacking trong dict literal

Cú pháp `**a` 'tháo' tất cả cặp key-value của `a` ra khỏi dict. Khi dùng trong `{}`:
```python
{**a, **b}
# tương đương:
# d = {}
# d.update(a)
# d.update(b)
```

Hữu ích khi cần tạo dict mới (immutable update pattern):
```python
config_default = {'lang': 'vi', 'theme': 'light', 'fps': 60}
config_user = {'theme': 'dark'}
config = {**config_default, **config_user}
# {'lang': 'vi', 'theme': 'dark', 'fps': 60}
```

### Vì sao 'bên phải thắng'?

Tự nhiên với cách Python xử lý gán: nếu key đã có và bạn gán giá trị mới, giá trị mới ghi đè. Tương tự, dict bên phải coi như được merge **sau** ⇒ thắng.

**Ứng dụng**: `merge(default, user)` — user override default, đúng kỳ vọng config.

````python
a = {'x': 1, 'y': 2}
b = {'y': 99, 'z': 3}

# Python 3.9+
print(a | b)            # {'x': 1, 'y': 99, 'z': 3}

# Cách cũ — chạy mọi phiên bản 3.x
print({**a, **b})

# Sửa a tại chỗ
a.update(b)
print(a)
````

### 3.5. Đảo dict — biến value thành key?

Cẩn thận: nếu value bị **trùng**, chỉ một key được giữ lại.

---

### Vì sao có thể bị mất dữ liệu?

Dict yêu cầu key **duy nhất**. Nếu 2 key trong dict gốc có cùng value, đảo lại:
```python
d = {'a': 1, 'b': 1}
{v: k for k, v in d.items()}
# {1: 'b'}  — key 'a' bị mất! 'b' xuất hiện sau nên thắng.
```

### Vì sao value cần hashable để làm key mới?

Như câu 2.9 — chỉ object **hashable** (immutable) mới làm key được. Nếu value là list:
```python
d = {'a': [1, 2], 'b': [3]}
{v: k for k, v in d.items()}
# TypeError: unhashable type: 'list'
```

Phải convert value thành tuple trước khi đảo: `{tuple(v): k for k, v in d.items()}`.

### Khi value trùng — gom thành list (giữ dữ liệu)

```python
from collections import defaultdict
gom = defaultdict(list)
for k, v in d.items():
    gom[v].append(k)   # mỗi value gom được nhiều key
```

Đây là pattern 'inverted index' — dùng nhiều trong search engines.

````python
d = {'a': 1, 'b': 2, 'c': 3}
dao = {v: k for k, v in d.items()}
print(dao)

# Khi value trùng — gom thành list:
from collections import defaultdict
d2 = {'a': 1, 'b': 2, 'c': 1, 'd': 2}
gom = defaultdict(list)
for k, v in d2.items():
    gom[v].append(k)
print(dict(gom))
````

### 3.6. Set — phép toán tập hợp cơ bản?

| Toán tử | Ý nghĩa |
|---------|---------|
| `A \| B` | hợp |
| `A & B` | giao |
| `A - B` | hiệu (chỉ ở A) |
| `A ^ B` | đối xứng (chỉ ở 1 trong 2) |
| `A <= B` | A là tập con của B |
| `A < B` | A là tập con thực sự |

![Venn diagrams - 4 phép toán tập hợp](../assets/images/appendix/set_operations.png)

---

### Vì sao Python chọn các ký hiệu này?

Lấy cảm hứng từ **toán học**:
- `∪` (union) ↔ `|` (cũng dùng cho 'or' bit-wise).
- `∩` (intersection) ↔ `&` (cũng dùng cho 'and' bit-wise).
- `\` (set difference trong toán) → đổi thành `-` (đẹp hơn trong code).
- `△` (symmetric difference) ↔ `^` (XOR — chỉ đúng nếu 'ở 1 trong 2 nhưng không cả 2').

### Hiệu năng — vì sao set nhanh?

Set dùng **hash table** giống dict:
- `x in s` ⇒ O(1) trung bình (vs O(n) cho list).
- `s.add(x)`, `s.remove(x)` ⇒ O(1) trung bình.
- `A & B` ⇒ O(min(|A|, |B|)) — duyệt set nhỏ hơn rồi check `in` set lớn hơn.

**Đánh đổi**: set không có thứ tự, không hỗ trợ index `s[0]`, phần tử phải hashable.

### Method-form vs operator-form

Mỗi toán tử có cả 2 dạng:

| Operator | Method | Khác biệt |
|----------|--------|-----------|
| `A \| B` | `A.union(B)` | Method nhận **iterable** bất kỳ (list, tuple, generator) |
| `A & B` | `A.intersection(B)` | tương tự |
| `A - B` | `A.difference(B)` | tương tự |
| `A ^ B` | `A.symmetric_difference(B)` | tương tự |

Vd: `s | [1,2,3]` ⇒ TypeError, nhưng `s.union([1,2,3])` OK.

### Mẹo: kiểm tra subset/superset

```python
users_active = {'lan', 'an', 'kha'}
vip = {'lan'}
vip <= users_active     # True — vip là subset
users_active >= vip     # True — users_active là superset
```
Hữu ích để check 'mọi quyền cần có đã được cấp'.

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print('Hợp:', A | B)
print('Giao:', A & B)
print('Hiệu:', A - B)
print('Đối xứng:', A ^ B)
print('A subset B?', A <= B)
````

### 3.7. Khi nào dùng `set` thay vì `list`?

Dùng `set` khi:
- Chỉ cần biết **có / không** (kiểm tra `in` siêu nhanh — O(1)).
- Cần **loại trùng**.
- Cần **phép toán tập hợp**.

**Đừng dùng set** khi cần thứ tự hoặc đếm số lần xuất hiện.

---

### Sự khác biệt O(1) vs O(n) — cụ thể

Code benchmark trong cell phía dưới minh hoạ điều này:
- Tìm `(n-1)` trong `list(range(n))` với n=1tr: **~50-100 ms** (linear scan).
- Tìm `(n-1)` trong `set(range(n))` với n=1tr: **~0.001 ms** (hash lookup).

Khoảng cách 10⁵ — đủ để biến code 'gần như đứng yên' thành 'tức thì'.

![list vs set lookup speed](../assets/images/appendix/list_vs_set.png)

### Quy tắc thực dụng

**Dùng set** nếu bạn:
- Sẽ kiểm tra `x in collection` **nhiều lần** (vd trong vòng lặp).
- Cần loại trùng (`set(my_list)` 1 dòng).
- Đang làm phép giao/hợp/hiệu giữa các tập.

**Dùng list** nếu bạn:
- Cần truy cập theo **thứ tự** hoặc **index** (`ds[0]`, `ds[-1]`).
- Cần **phần tử trùng** (vd điểm số 4 lần được 9.0).
- Phần tử **không hashable** (list của list, list của dict).

### Bẫy: set không deterministic giữa các lần chạy

Trong Python 3.3+, **hash của str có random salt** mỗi lần khởi động Python → thứ tự duyệt set có thể đổi. Đừng phụ thuộc vào thứ tự này (kể cả khi 'có vẻ' nhất quán).

````python
# So sánh tốc độ:
import time
n = 1_000_000
data_list = list(range(n))
data_set = set(data_list)

t0 = time.perf_counter()
_ = (n - 1) in data_list
print(f'list `in`: {time.perf_counter() - t0:.5f}s')

t0 = time.perf_counter()
_ = (n - 1) in data_set
print(f'set  `in`: {time.perf_counter() - t0:.5f}s')
````

## 4. Lambda, map, filter, reduce _( Lambda, map, filter, reduce )_


Đây là **lập trình hàm (functional programming)** trong Python. Hữu ích khi làm việc với dữ liệu — nhưng hãy nhớ: trong nhiều trường hợp, **list comprehension dễ đọc hơn**.

### 4.1. `lambda` là gì? Khi nào nên dùng?

**Lambda** là **hàm 1 dòng không tên** (anonymous function).

Cú pháp: `lambda tham_số: biểu_thức`.

Tương đương:
```python
binh_phuong = lambda x: x ** 2
# giống như:
def binh_phuong(x):
    return x ** 2
```

**Khi nào dùng:** khi cần một hàm rất ngắn để truyền vào `sorted(key=)`, `map`, `filter`, ... và sẽ không tái sử dụng.

**Khi nào KHÔNG dùng:** nếu logic > 1 dòng, viết `def` đàng hoàng.

---

### Phân tích cú pháp `lambda`

```
lambda  tham_số_1, tham_số_2, ...:  biểu_thức
─┬───  ─┬─────────────────────────  ─┬────────
 │      │                            │
 │      │                            └── giá trị **trả về** (chỉ 1 biểu thức)
 │      └── danh sách tham số (giống def)
 └── từ khoá lambda
```

Lambda **không có** `return` — biểu thức cuối tự động được trả về.

### Hạn chế của lambda — vì sao?

Lambda **chỉ chứa 1 biểu thức** — không được:
- Câu lệnh nhiều dòng (vd vòng lặp, gán biến).
- `if/else dạng câu lệnh` (nhưng **biểu thức ternary** thì OK: `lambda x: 'odd' if x % 2 else 'even'`).
- `return`, `yield`, `try/except`.

Lý do triết lý: tác giả Guido (creator của Python) muốn lambda **ngắn và đơn giản**. Nếu cần nhiều, hãy viết `def`.

### Lambda với mặc định, *args, **kwargs

Mọi tính năng của def đều có:
```python
f = lambda x, y=10, *args, **kwargs: x + y + sum(args) + sum(kwargs.values())
f(1, 2, 3, 4, a=5)   # 1 + 2 + 7 + 5 = 15
```

### Lambda đã đặt tên — anti-pattern

PEP 8 (style guide chính thức) **không khuyến khích** gán lambda cho biến:
```python
# Tệ — đã có tên thì viết def cho rõ
binh_phuong = lambda x: x ** 2

# Tốt
def binh_phuong(x):
    return x ** 2
```

Vì sao? Khi báo lỗi traceback của hàm đầu sẽ in `<lambda>` (không có tên), rất khó debug. Hàm thường in tên rõ ràng.

````python
binh_phuong = lambda x: x ** 2
print(binh_phuong(5))

# Dùng với sorted:
ten = ['Châu', 'An', 'Bình']
print(sorted(ten, key=lambda s: len(s)))

# Lambda nhiều tham số:
cong = lambda a, b: a + b
print(cong(3, 4))
````

> Lambda **chỉ chứa 1 biểu thức** — không được có `if/else` dạng câu lệnh, không được có vòng lặp. Nếu cần, viết `def`.

### 4.2. `map()` — biến đổi từng phần tử?

`map(fn, iterable)` áp dụng `fn` cho từng phần tử, trả về iterator. Cần `list()` để in.

**So sánh với list comprehension** — comprehension thường rõ hơn.

---

### Vì sao `map` trả iterator chứ không trả list?

Trong Python 3, `map` được thiết kế **lazy** — không tính ngay mà chỉ khi cần:

```python
ds = [1, 2, 3, 4, 5]
kq = map(lambda x: x ** 2, ds)
print(kq)             # <map object at 0x...> — chưa tính
print(list(kq))       # [1, 4, 9, 16, 25] — tính lúc này
```

Lợi ích: nếu data lớn (1tr phần tử) và bạn chỉ cần 10 phần tử đầu, map không lãng phí RAM tính cả 1tr — chỉ tính từng cái khi duyệt.

**Bẫy**: iterator chỉ duyệt **được 1 lần**. Sau khi `list(kq)`, lần sau `list(kq)` ra `[]`.

### `map` với 2+ iterables

Nếu `fn` có 2 tham số, truyền 2 iterables — `map` ghép cặp như `zip`:
```python
a = [1, 2, 3]
b = [10, 20, 30]
list(map(lambda x, y: x + y, a, b))   # [11, 22, 33]
```

Dừng ở iterable **ngắn nhất** (giống zip).

### Tại sao comprehension thường được ưa hơn?

Comprehension đọc tự nhiên như tiếng Anh:
```python
[x ** 2 for x in ds]      # 'bình phương x cho mỗi x trong ds'
list(map(lambda x: x ** 2, ds))   # cùng kết quả nhưng đọc rườm rà hơn
```

Comprehension còn cho phép thêm `if` dễ dàng:
```python
[x ** 2 for x in ds if x > 0]   # comprehension: gọn
list(map(lambda x: x ** 2, filter(lambda x: x > 0, ds)))   # 2 lambdas — rối
```

**Nhưng `map` thắng** khi:
- Hàm đã có sẵn tên (không cần lambda): `list(map(int, strings))` ngắn hơn `[int(s) for s in strings]`.
- Cần **lazy** thực sự (data cực lớn).


````python
ds = [1, 2, 3, 4, 5]

# map
binh = list(map(lambda x: x ** 2, ds))
print(binh)

# Tương đương với comprehension (Pythonic hơn):
print([x ** 2 for x in ds])

# map với 2 iterable — cộng từng cặp:
a = [1, 2, 3]
b = [10, 20, 30]
print(list(map(lambda x, y: x + y, a, b)))
````

### 4.3. `filter()` — lọc theo điều kiện?

`filter(predicate, iterable)` giữ lại các phần tử khiến `predicate(x)` là `True`.

---

### Predicate là gì?

**Predicate** = hàm trả về `True`/`False`. Các tên này tương đương trong toán/logic.

`filter(pred, ds)` duyệt qua từng `x` của `ds`, **giữ lại chỉ những `x` mà `pred(x)` là truthy**.

### `filter(None, iterable)` — trick lọc falsy

Nếu truyền `None` làm predicate, `filter` lọc các giá trị **truthy** (xem 6.3 về truthy/falsy):
```python
ds = [0, 1, '', 'hi', None, [], [0], False, True]
list(filter(None, ds))   # [1, 'hi', [0], True]
```

Tương đương `[x for x in ds if x]`.

### Lazy giống map

`filter` cũng trả iterator — không tính ngay. Lý do và bẫy giống `map`.

### Khi nào filter > comprehension?

Filter ngắn hơn khi predicate là **hàm có sẵn tên**:
```python
list(filter(str.isdigit, ['1', 'a', '2', 'b']))   # filter: ngắn
[s for s in ['1','a','2','b'] if s.isdigit()]      # comp: tương đương
```


````python
ds = [3, 8, 1, 9, 4, 2, 7]

chan = list(filter(lambda x: x % 2 == 0, ds))
print(chan)

# Tương đương:
print([x for x in ds if x % 2 == 0])
````

### 4.4. `reduce()` — gộp dồn về 1 giá trị?

`reduce(fn, iterable, initial)` áp dụng `fn` từ trái sang phải, kết quả của bước trước là input của bước sau. Phải `from functools import reduce`.

---

### Cách reduce hoạt động — bước-by-bước

Với `reduce(f, [a, b, c, d])`, Python tính:
1. `acc = a`
2. `acc = f(acc, b)`
3. `acc = f(acc, c)`
4. `acc = f(acc, d)`
5. Trả `acc`.

Vd `reduce(lambda a, b: a + b, [1, 2, 3, 4])`:
- acc = 1
- acc = 1 + 2 = 3
- acc = 3 + 3 = 6
- acc = 6 + 4 = 10

### Tham số `initial` — giá trị khởi tạo

`reduce(f, ds, init)` tương đương `reduce(f, [init] + ds)`.

Hữu ích khi:
- List rỗng cần fallback (vd `reduce(max, [])` raise lỗi, `reduce(max, [], 0)` trả 0).
- Bắt đầu từ giá trị không có trong list (vd reduce 1 list của list thành list lớn: `reduce(lambda a, b: a + b, lists, [])`).

### Vì sao Python ít dùng reduce?

Guido van Rossum (creator) đã có lần nói reduce 'thường khó đọc hơn vòng lặp thường'. Trong Python 3, reduce bị **chuyển vào `functools`** (không còn built-in) để khuyến khích dùng:
- `sum()`, `max()`, `min()`, `any()`, `all()` — đã đủ cho 90% trường hợp.
- Vòng `for` thường rõ ý hơn cho các trường hợp lạ.

**Khi reduce thực sự sáng giá**: tính toán có **trạng thái cumulative** mà không có built-in nào tương đương:
```python
from functools import reduce
# Tích tổ hợp: gộp nhiều dict thành 1
reduce(lambda a, b: {**a, **b}, list_of_dicts, {})
```

### `accumulate` — anh em của reduce

Khác biệt: reduce trả **1 giá trị cuối**, accumulate trả **list các bước trung gian**:
```python
from itertools import accumulate
list(accumulate([1, 2, 3, 4]))         # [1, 3, 6, 10] — tích luỹ tổng
list(accumulate([1, 2, 3, 4], max))    # [1, 2, 3, 4] — running max
```


````python
from functools import reduce

ds = [1, 2, 3, 4, 5]

# Tổng:
print(reduce(lambda a, b: a + b, ds))
print(sum(ds))   # tương đương — dùng sum cho gọn

# Tích:
print(reduce(lambda a, b: a * b, ds))

# Tìm max thủ công:
print(reduce(lambda a, b: a if a > b else b, ds))
````

> Thực tế: `sum`, `max`, `min`, `any`, `all` đã đủ cho 90% trường hợp. `reduce` ít dùng.

### 4.5. Lambda vs comprehension — khi nào dùng cái nào?

**Comprehension** rõ ràng hơn cho 'biến đổi list'.  
**Lambda** thường dùng khi cần **truyền hàm** cho một hàm khác (`sorted`, `max`, `min`, ...).

| Tình huống | Nên dùng |
|------------|----------|
| Tạo list mới từ list cũ | Comprehension |
| Lọc list | Comprehension `[x for x in ds if ...]` |
| Tham số `key=` của sort/max/min | Lambda |
| Logic > 1 biểu thức | `def` |

---

### 2 phong cách lập trình khác nhau

- **Functional** (lambda + map/filter/reduce): **mô tả phép biến đổi**, không quan tâm chi tiết.
- **Imperative** (vòng for + if): **mô tả từng bước**, kiểm soát trực tiếp.

Python được thiết kế **hybrid** — hỗ trợ cả 2, nhưng **comprehension** (còn gọi là 'comprehension form') thường được coi là 'lập trình tự nhiên' nhất trong Python.

### Quy tắc thực dụng — chọn cái nào trong 5 giây?

Đọc thử 2 cách viết và hỏi: **'Cách nào tôi đọc thấy ý đồ ngay lập tức?'**

```python
# Bài toán: lấy độ dài các từ dài hơn 3 ký tự, viết hoa
tu = ['hi', 'python', 'go', 'java']

# Comprehension — đọc tự nhiên
[t.upper() for t in tu if len(t) > 3]

# map + filter + lambda — rườm rà với 2 lambdas
list(map(lambda t: t.upper(), filter(lambda t: len(t) > 3, tu)))
```

Comprehension thắng rõ rệt.

Nhưng:
```python
# Bài toán: sort theo độ dài giảm dần
sorted(tu, key=lambda t: -len(t))   # lambda thắng — comp không phù hợp
```

### Một số trường hợp lambda đẹp hơn comp

1. **Tham số `key=`** của `sorted/max/min/groupby/...`:
   ```python
   max(hs, key=lambda h: h['diem'])
   ```
2. **Truyền callback** cho thư viện (matplotlib, pandas apply, ...):
   ```python
   df['ten'].apply(lambda s: s.title())
   ```
3. **One-liner trong test/REPL**: nhanh, không cần đặt tên.


````python
# Bài toán: sắp các từ theo độ dài giảm dần
tu = ['cherry', 'apple', 'banana', 'fig', 'date']

# Cách Pythonic: lambda cho key
print(sorted(tu, key=lambda s: len(s), reverse=True))

# Bài toán: lấy bình phương của các số chẵn
ds = list(range(10))

# Comprehension — gọn nhất:
print([x ** 2 for x in ds if x % 2 == 0])

# map + filter — không Pythonic bằng:
print(list(map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, ds))))
````

## 5. Comprehensions nâng cao _( Advanced Comprehensions )_


Comprehensions là **đặc sản Python**. Học cách dùng đúng giúp code ngắn và đẹp. Học cách dùng sai sẽ tạo ra code không ai đọc nổi — kể cả chính bạn sau 1 tuần.

### 5.1. Có if-else trong comprehension như thế nào?

Phân biệt 2 trường hợp:

**Lọc** (`if` ở cuối):
```python
[x for x in ds if x > 0]
```

**Biến đổi có điều kiện** (`if-else` trong biểu thức):
```python
[x if x >= 0 else -x for x in ds]
```

Có thể kết hợp cả hai.

![Hai vị trí của if trong comprehension](../assets/images/appendix/comprehension_positions.png)

---

### Vì sao 2 vị trí khác nhau lại có ý nghĩa khác nhau?

Cú pháp đầy đủ của comprehension:

```
[ <biểu_thức>  for <biến> in <iterable>  if <điều_kiện_lọc> ]
    ▲                                          ▲
    └── Nơi if-else dạng biểu thức              └── Nơi if dạng lọc (chỉ if, KHÔNG có else)
```

- **Vị trí 1 (biểu thức)**: `if-else` ở đây tính giá trị → cho **mọi** phần tử. Phải có `else`.
- **Vị trí 2 (lọc)**: `if` ở đây quyết định giữ hay bỏ phần tử → giảm số lượng đầu ra. **Không** dùng `else` được.

### Bẫy syntax: nhầm 2 vị trí

```python
# ❌ SAI — `if-else` ở cuối là lỗi cú pháp
[x for x in ds if x > 0 else 0]

# ✓ ĐÚNG — `if-else` ở đầu (biểu thức), không có lọc
[x if x > 0 else 0 for x in ds]

# ✓ ĐÚNG — `if` ở cuối (lọc), không có else
[x for x in ds if x > 0]
```

### Kết hợp cả hai

```python
# Lấy abs(x) cho x != 0, và bỏ qua x == 0
[x if x > 0 else -x for x in ds if x != 0]
```

Đọc theo thứ tự:
1. Duyệt `for x in ds`.
2. Lọc `if x != 0` (giữ lại x ≠ 0).
3. Tính giá trị `x if x > 0 else -x` cho từng x đã giữ.


````python
ds = [-3, 5, -2, 8, 0, -7]

# Lọc số dương:
print([x for x in ds if x > 0])

# Lấy giá trị tuyệt đối:
print([x if x >= 0 else -x for x in ds])

# Kết hợp: lấy abs nhưng bỏ số 0:
print([x if x > 0 else -x for x in ds if x != 0])
````

### 5.2. Comprehension lồng (nested) — khi nào dùng, khi nào không?

**Nên dùng** cho các thao tác đơn giản như flatten 2D hay tạo ma trận.

**Đừng dùng** nếu có 3 cấp lồng hoặc logic phức tạp — chuyển sang vòng `for` thường.

---

### 2 dạng comprehension lồng — KHÁC NHAU

**Dạng 1 — Flatten (nhiều `for` trong cùng 1 cặp `[]`):**
```python
[x for hang in M for x in hang]
```
Đọc: 'lấy x, với hang trong M, với x trong hang' → **làm phẳng** thành 1 list.

**Dạng 2 — Build matrix (comprehension lồng comprehension):**
```python
[[i + j for j in range(3)] for i in range(3)]
```
Đọc: 'với mỗi i, tạo một list `[i+j for j ...]`' → **list của list**, vẫn 2 chiều.

Khác biệt nằm ở **dấu `[` thứ 2**: có hay không.

### Vì sao thứ tự `for` lại quan trọng?

```python
M = [[1, 2], [3, 4]]

# Cách A: 'hang trước, x sau'
[x for hang in M for x in hang]    # [1, 2, 3, 4] — đúng

# Cách B: đảo thứ tự
[x for x in hang for hang in M]    # NameError: hang chưa được định nghĩa
```

Quy tắc: viết comprehension **giống vòng for thường** (vòng ngoài → vòng trong → biểu thức), rồi 'kéo biểu thức lên đầu'.

### Khi nào KHÔNG dùng comprehension?

Quy tắc 3 dòng: **nếu cần > 3 dòng comment để giải thích**, đổi sang vòng for thường.

Code dưới đây 'đúng nhưng không ai đọc nổi':
```python
kq = [[x*y for y in range(1, 11) if y % 2 == 0] for x in range(2, 10) if x not in (5, 7)]
```

Tách thành for thường rõ hơn rất nhiều.

````python
# Flatten 2D — ổn:
M = [[1, 2, 3], [4, 5], [6]]
print([x for hang in M for x in hang])

# Tạo ma trận 3x3 toạ độ:
grid = [[(i, j) for j in range(3)] for i in range(3)]
for hang in grid:
    print(hang)

# Cặp khác nhau từ 2 list:
a = [1, 2, 3]
b = ['x', 'y']
print([(i, j) for i in a for j in b])
````

> Quy tắc: nếu cần > 2 dòng comment để giải thích comprehension → đổi sang for thường.

### 5.3. Dict comprehension và set comprehension?

Cú pháp tương tự list, chỉ đổi ngoặc và format:

```python
# dict
{key_expr: value_expr for x in iterable}

# set
{expr for x in iterable}
```

---

### Phân biệt set comp và dict comp — qua dấu `:`

Cả 2 đều dùng `{ }` — Python phân biệt qua việc có dấu `:` hay không:

```python
{x for x in range(5)}        # set: {0, 1, 2, 3, 4}
{x: x*2 for x in range(5)}   # dict: {0: 0, 1: 2, 2: 4, 3: 6, 4: 8}
```

### Bẫy: dict literal rỗng `{}` là **dict**, không phải set!

```python
type({})       # <class 'dict'> — RỖNG → mặc định dict
type(set())    # <class 'set'> — phải dùng set() để tạo set rỗng
type({1})      # <class 'set'>
```

Lý do: dict được giới thiệu trước set — nên `{}` 'thuộc về' dict.

### Trick: tạo dict từ 2 list

Cú pháp Pythonic dùng `zip`:
```python
ten = ['An', 'Bình', 'Châu']
diem = [8.5, 7.0, 9.0]

{t: d for t, d in zip(ten, diem)}      # comprehension
dict(zip(ten, diem))                    # ngắn hơn, cùng kết quả
```

### Set comp lọc trùng + biến đổi cùng lúc

```python
# Lấy các nguyên âm KHÁC NHAU trong câu
{c.lower() for c in 'Học Python rất vui' if c.lower() in 'aeiou'}
# {'o', 'u', 'i'}
```
Set tự loại trùng — không cần check `not in seen`.

````python
# Dict comprehension: bình phương của 1..5
binh = {x: x ** 2 for x in range(1, 6)}
print(binh)

# Đảo dict (key ↔ value):
d = {'a': 1, 'b': 2, 'c': 3}
print({v: k for k, v in d.items()})

# Set comprehension: các nguyên âm trong câu
ng = {c for c in 'Học Python rất vui' if c in 'aeiou'}
print(ng)
````

### 5.4. Generator expression vs list comprehension — khác gì?

Đổi `[ ]` thành `( )` → **generator expression**.  
Khác biệt:
- List comp **tạo cả list ngay** trong bộ nhớ.
- Generator expression **lười (lazy)** — chỉ sinh giá trị khi cần.

Khi nào dùng generator?
- Khi data lớn, không cần giữ tất cả trong RAM.
- Khi truyền vào hàm aggregate (`sum`, `max`, `any`, `all`) — không cần `[ ]`.

---

### Lazy là gì? Vì sao tiết kiệm RAM?

```python
lst = [x ** 2 for x in range(10**6)]   # tính NGAY 1 triệu phần tử → ~30 MB RAM
gen = (x ** 2 for x in range(10**6))   # KHÔNG tính — chỉ giữ 'công thức' → ~200 bytes

for v in gen:    # tính từng cái khi duyệt — RAM tối thiểu
    if v > 1000:
        break    # dừng sớm — chỉ tính ~32 phần tử thay vì 1 triệu
```

### Bẫy: generator chỉ duyệt được 1 lần

```python
gen = (x for x in range(5))
list(gen)    # [0, 1, 2, 3, 4]
list(gen)    # [] — generator đã 'cạn'
```

Nếu cần duyệt nhiều lần, dùng list comp.

### Trick: bỏ ngoặc khi truyền vào hàm 1 tham số

```python
sum((x ** 2 for x in range(10)))   # OK
sum(x ** 2 for x in range(10))     # cũng OK — Python 'mượn' ngoặc của hàm
```

Pythonic — đặc biệt với `sum`, `max`, `min`, `any`, `all`.

### Hỗn hợp: list comp **bên trong** generator?

```python
max([sum(row) for row in matrix])    # tạo list 1tr phần tử rồi tìm max
max(sum(row) for row in matrix)      # lazy — RAM cố định
```

Khi chỉ cần 1 giá trị aggregate, dùng generator → tiết kiệm RAM.

### Khi nào list > generator?

- Cần **truy cập index** (`ds[3]`, `ds[-1]`).
- Cần **`len()`**.
- Cần **duyệt nhiều lần**.
- Data nhỏ (vài trăm phần tử): khác biệt RAM không đáng kể, list rõ hơn.

````python
# So sánh:
import sys
lst = [x ** 2 for x in range(1000)]
gen = (x ** 2 for x in range(1000))
print('list size:', sys.getsizeof(lst), 'bytes')
print('gen  size:', sys.getsizeof(gen), 'bytes')

# Pythonic — bỏ luôn ngoặc khi truyền vào sum/max/...:
print(sum(x ** 2 for x in range(1, 11)))
print(max(len(s) for s in ['táo', 'cam', 'chuối']))
````

## 6. Idioms Python phổ biến _( Common Python Idioms )_


**Idiom** là cách diễn đạt 'rất Python' cho các thao tác thường gặp. Học chúng để code đọc tự nhiên hơn.

### 6.1. Đổi giá trị 2 biến (swap) — không cần biến tạm?

Python cho phép `a, b = b, a` nhờ tuple packing/unpacking.

---

### Cơ chế: chuyện gì xảy ra phía sau `a, b = b, a`?

Python đánh giá biểu thức **bên phải** trước, **rồi** gán vào bên trái:

1. **Đánh giá vế phải**: `b, a` → tạo tuple `(b_value, a_value)` (packing).
2. **Unpacking** vế trái: `a` nhận phần tử 0 = `b_value`, `b` nhận phần tử 1 = `a_value`.

Nhờ bước 1 'chụp ảnh' giá trị cũ trước, không cần biến tạm.

### Tổng quát hoá — gán nhiều biến

Cú pháp `a, b, c = x, y, z` hoạt động cho mọi số biến:

```python
a, b, c = 1, 2, 3
a, b, c = c, a, b   # rotate: (3, 1, 2)

# Unpack từ list/tuple/string:
x, y, z = [10, 20, 30]
a, b = 'hi'             # 'h', 'i'
```

### `*rest` — gom phần còn lại

```python
first, *middle, last = [1, 2, 3, 4, 5]
# first=1, middle=[2, 3, 4], last=5

head, *tail = [1, 2, 3, 4]
# head=1, tail=[2, 3, 4]
```

Hữu ích: chia 'phần tử đầu / cuối' và 'phần còn lại' rõ nét.

### So với cách C/Java thủ công

Cách truyền thống (vẫn hoạt động trong Python):
```python
tmp = a
a = b
b = tmp
```

Python's `a, b = b, a`:
- Ngắn hơn 3 lần.
- Không tốn biến tạm có tên (đỡ ô nhiễm namespace).
- Rõ ý hơn — đọc 1 lèo là hiểu.

````python
a = 1
b = 2
a, b = b, a
print(a, b)

# Còn dùng được cho 3 biến:
a, b, c = 1, 2, 3
a, b, c = c, a, b
print(a, b, c)
````

### 6.2. Toán tử ba ngôi (ternary) — `if-else` trên một dòng?

Cú pháp: `value_if_true if condition else value_if_false`.

Đọc từ trái sang phải, đảo so với một số ngôn ngữ khác (`cond ? a : b`).

---

### Vì sao Python chọn cú pháp khác đa số ngôn ngữ?

Đa số ngôn ngữ (C, Java, JS, ...) dùng `cond ? a : b` — đọc 'NẾU cond THÌ a NGƯỢC LẠI b'.

Python chọn `a if cond else b` — đọc 'a NẾU cond NGƯỢC LẠI b'.

Lý do: Python ưu tiên **cách đọc tự nhiên trong tiếng Anh** — câu hay viết kiểu 'I'll have **tea if it's hot else** coffee'. Giá trị thường thấy đứng trước, điều kiện đứng sau.

### Phân biệt **biểu thức** vs **câu lệnh**

- **Câu lệnh `if`**: `if cond: ...` — chỉ làm gì đó, không trả giá trị.
- **Biểu thức ternary**: `a if cond else b` — **trả về giá trị**, có thể gán hoặc đặt vào biểu thức lớn hơn.

```python
# Câu lệnh — không gán được
if x > 0:
    sign = 1
else:
    sign = -1

# Biểu thức — 1 dòng
sign = 1 if x > 0 else -1

# Trong list comp
[1 if x > 0 else -1 for x in ds]

# Trong f-string
f'Số {x} là {"dương" if x > 0 else "âm hoặc 0"}'
```

### Bẫy: ternary lồng — đọc khó

```python
loai = 'Giỏi' if d >= 8 else 'Khá' if d >= 6.5 else 'TB' if d >= 5 else 'Yếu'
```

Đọc đúng: `(((Giỏi if d>=8 else Khá) if d>=6.5 else TB) if d>=5 else Yếu)` — right-associative. Quá 2 cấp lồng → đổi sang if/elif/else thường.

### Một số bẫy khác

- `a if cond` (không có else) → SyntaxError. Ternary **bắt buộc có else**.
- `1 if False else 0 + 5` → kết quả `5`, vì `+` ưu tiên cao hơn `if-else` → thực ra là `1 if False else (0 + 5)`.

````python
x = 7
loai = 'chẵn' if x % 2 == 0 else 'lẻ'
print(loai)

# Lồng — tránh nếu được:
diem = 7.5
loai = 'Giỏi' if diem >= 8 else ('Khá' if diem >= 6.5 else 'TB')
print(loai)
````

### 6.3. Truthy / Falsy là gì?

Trong điều kiện `if`, Python coi nhiều giá trị **khác `False` cũng là 'sai'**:

**Falsy:** `False`, `None`, `0`, `0.0`, `''`, `[]`, `()`, `{}`, `set()`.

**Truthy:** mọi thứ khác (kể cả chuỗi `'0'`, list `[0]`, ...).

---

### Vì sao `[]` lại falsy?

Quy tắc Python: collection **rỗng** (không có phần tử nào) → falsy. Lý do triết lý: 'rỗng = không có gì = sai/không có'.

Cụ thể với từng kiểu:
- `[]`, `()`, `{}`, `set()`, `''` → falsy (collection rỗng).
- `[0]`, `(False,)`, `{None: 0}`, `'0'` → **truthy** (có phần tử, dù phần tử đó falsy).

### Bẫy phổ biến

**Bẫy 1**: kiểm tra 'biến đã được set chưa'
```python
tuoi = 0   # tuổi 0 vẫn hợp lệ
if tuoi:   # ❌ False — bỏ qua giá trị 0 hợp lệ
    ...
if tuoi is not None:   # ✓ đúng ý 'đã được gán'
    ...
```

**Bẫy 2**: hàm trả về list rỗng vs `None`
```python
kq = ham_co_the_tra_list()    # có thể trả [] hoặc None
if kq:                        # cả [] và None đều falsy → cùng nhánh
    ...
if kq is not None:            # phân biệt 'không có' vs 'rỗng'
    ...
```

### Idiom hữu ích

**Default value với `or`:**
```python
ten = input_ten or 'Khách'    # nếu input_ten là '' → 'Khách'
```

Hoạt động vì `or` trả về **giá trị**, không phải True/False:
- `a or b` → `a` nếu truthy, ngược lại `b`.
- `a and b` → `b` nếu cả 2 truthy, ngược lại trả về cái đầu falsy.

**Bẫy**: `0 or 5 → 5` (không phải 0), nên cẩn thận khi default value có thể là 0.

### `bool(x)` để kiểm tra rõ ràng

Khi cần một giá trị truthy/falsy cụ thể (không phải kiểu gốc):
```python
bool([])      # False
bool([0])     # True
bool('')      # False
bool(' ')     # True (có khoảng trắng → không rỗng)
```


````python
for v in [0, 1, '', 'hi', [], [0], None, False, {}]:
    print(repr(v), '->', bool(v))

# Idiom phổ biến — kiểm tra list rỗng:
ds = []
if not ds:
    print('list rỗng')

# Default value với 'or':
ten = '' or 'Không tên'
print(ten)
````

### 6.4. `enumerate` và `zip` — duyệt index + nhiều list cùng lúc?

- `enumerate(iter, start=0)` → cặp `(index, value)`.
- `zip(a, b, c, ...)` → cặp `(a[i], b[i], c[i])` cho đến khi list ngắn nhất hết.

---

### Vì sao `enumerate` thay cho `range(len())`?

Cách thủ công:
```python
for i in range(len(ds)):
    print(i, ds[i])
```

Cách Pythonic:
```python
for i, x in enumerate(ds):
    print(i, x)
```

Cách 2 tốt hơn vì:
- Không cần truy cập `ds[i]` (1 phép toán phụ).
- Hoạt động với mọi iterable (kể cả generator không có `len`).
- Đọc tự nhiên: 'với mỗi (index, value) trong ds'.

### Tham số `start=` của `enumerate`

```python
for i, x in enumerate(['a', 'b', 'c'], start=1):
    print(f'{i}. {x}')   # 1. a, 2. b, 3. c
```

Hữu ích để in danh sách bắt đầu từ 1 (cho người đọc) thay vì 0.

### `zip` dừng ở list ngắn nhất — bẫy phổ biến

```python
a = [1, 2, 3, 4, 5]
b = [10, 20]
list(zip(a, b))   # [(1, 10), (2, 20)] — bỏ 3 phần tử của a!
```

**Im lặng nuốt dữ liệu** → bug khó phát hiện.

**Cách 1**: dùng `zip_longest` từ `itertools`:
```python
from itertools import zip_longest
list(zip_longest(a, b, fillvalue=0))
# [(1, 10), (2, 20), (3, 0), (4, 0), (5, 0)]
```

**Cách 2** (Python 3.10+): `zip(a, b, strict=True)` raise lỗi nếu độ dài khác:
```python
list(zip(a, b, strict=True))   # ValueError: zip() argument 2 is shorter
```

### Unzip — đảo ngược zip

Trick `zip(*list_of_tuples)` tách 1 list các tuple thành các list riêng:
```python
cap = [(1, 'a'), (2, 'b'), (3, 'c')]
so, chu = zip(*cap)
# so = (1, 2, 3), chu = ('a', 'b', 'c')
```

Hoạt động vì `zip(*x)` tương đương `zip(x[0], x[1], x[2])` — mỗi tuple thành 1 'cột'.

````python
ten = ['An', 'Bình', 'Châu']
diem = [8, 7, 9]

# enumerate — đánh số:
for i, t in enumerate(ten, start=1):
    print(f'{i}. {t}')

# zip — duyệt 2 list cùng lúc:
for t, d in zip(ten, diem):
    print(t, d)

# Tạo dict từ 2 list:
print(dict(zip(ten, diem)))

# Unzip:
cap = [(1, 'a'), (2, 'b'), (3, 'c')]
so, chu = zip(*cap)
print(so, chu)
````

### 6.5. `*args` và `**kwargs` — hàm nhận số tham số tuỳ ý?

- `*args` — gói các đối số dương thành **tuple**.
- `**kwargs` — gói các đối số tên thành **dict**.

Tên `args` và `kwargs` chỉ là quy ước — có thể đặt khác.

---

### Phân biệt **đối số dương** (positional) vs **đối số tên** (keyword)

```python
def f(a, b, c):
    ...

f(1, 2, 3)            # 3 đối số dương — theo thứ tự
f(1, b=2, c=3)        # 1 dương + 2 tên
f(a=1, b=2, c=3)      # 3 đối số tên
```

`*args` gói **đối số dương dư**, `**kwargs` gói **đối số tên dư**.

### Cú pháp `*` và `**` — 2 vai trò khác nhau

**Trong định nghĩa hàm**: gói tham số.
```python
def f(*args, **kwargs):
    print(args)      # tuple
    print(kwargs)    # dict

f(1, 2, x=10)
# args = (1, 2)
# kwargs = {'x': 10}
```

**Khi gọi hàm**: tháo (unpack) iterable / dict thành đối số.
```python
def f(a, b, c):
    print(a, b, c)

args = [1, 2, 3]
f(*args)              # tương đương f(1, 2, 3)

kwargs = {'a': 1, 'b': 2, 'c': 3}
f(**kwargs)           # tương đương f(a=1, b=2, c=3)
```

Đối xứng đẹp: dùng `*` để gói khi định nghĩa, dùng `*` để tháo khi gọi.

### Pattern phổ biến: forward arguments

```python
def log_va_goi(fn, *args, **kwargs):
    print(f'Gọi {fn.__name__}({args}, {kwargs})')
    return fn(*args, **kwargs)   # tháo ra để gọi fn

log_va_goi(max, 1, 2, 3)
log_va_goi(sorted, [3, 1, 2], reverse=True)
```

Đây là cách viết **decorator** và **wrapper function** — cực kỳ phổ biến.

### Bẫy: thứ tự tham số

Trong `def`, thứ tự bắt buộc:
```python
def f(thuong, mac_dinh=10, *args, keyword_only, **kwargs):
    ...
```

Đối số sau `*args` là **keyword-only** — phải gọi bằng tên.

````python
def cong_tat_ca(*args):
    return sum(args)

print(cong_tat_ca(1, 2, 3, 4, 5))

def in_thong_tin(**kwargs):
    for k, v in kwargs.items():
        print(f'{k} = {v}')

in_thong_tin(ten='An', tuoi=16, lop='10A1')

# Cả hai — kết hợp với tham số bình thường:
def f(a, b, *args, **kwargs):
    print(a, b, args, kwargs)

f(1, 2, 3, 4, x=5, y=6)
````

### 6.6. Toán tử `:=` (walrus) — gán và dùng giá trị trong cùng biểu thức?

Python 3.8+. Cú pháp `name := expr` gán **và** trả giá trị. Dùng khi muốn gọi một thứ tốn kém **một lần** rồi vừa kiểm tra vừa lưu.

---

### Vì sao gọi là 'walrus'?

Vì `:=` trông giống **mặt con hải mã (walrus)** — 2 chấm là mắt, dấu `=` là ngà. Một sự đùa của cộng đồng Python.

### Phân biệt `=` và `:=`

- `=` là **câu lệnh** (statement) — chỉ đứng riêng dòng, không trả giá trị.
- `:=` là **biểu thức** (expression) — vừa gán vừa trả giá trị, có thể nhúng vào biểu thức khác.

```python
x = 5            # OK — câu lệnh
if x = 5:        # ❌ SyntaxError — không thể dùng = trong if
if (x := 5):     # ✓ OK — gán x=5 và dùng giá trị 5 (truthy)
```

### Use case 1: gọi hàm tốn kém 1 lần

**Trước walrus:**
```python
kq = ham_tinh_lau()
if kq is not None:
    process(kq)
```

**Với walrus:**
```python
if (kq := ham_tinh_lau()) is not None:
    process(kq)
```

Gọn hơn, đỡ tách thành 2 dòng.

### Use case 2: vòng lặp đọc input

**Trước walrus:**
```python
while True:
    line = input()
    if line == 'quit':
        break
    process(line)
```

**Với walrus:**
```python
while (line := input()) != 'quit':
    process(line)
```

### Use case 3: comprehension không lặp tính toán

**Tệ — `len(s)` tính 2 lần:**
```python
[(s, len(s)) for s in tu if len(s) > 3]
```

**Walrus — chỉ tính 1 lần:**
```python
[(s, n) for s in tu if (n := len(s)) > 3]
```

### Khi nào KHÔNG dùng walrus?

- Code đơn giản đã đủ → đừng tăng thêm cú pháp lạ.
- Người đọc chưa quen Python 3.8+ → có thể bối rối.
- Khi chỉ cần gán đơn thuần → dùng `=` cho rõ.

Quy tắc: chỉ dùng walrus khi nó **rút gọn rõ rệt** code (1-2 dòng).

### Ưu tiên (precedence)

Walrus có **ưu tiên thấp** — nên thường cần ngoặc:
```python
if x := f() and x > 0:    # SAI — đánh giá x := (f() and x > 0)
if (x := f()) and x > 0:  # ĐÚNG
```


````python
# Trước:
ds = [1, 2, 3, 4, 5]
n = len(ds)
if n > 3:
    print(f'List có {n} phần tử (>3)')

# Walrus — gọn hơn:
if (n := len(ds)) > 3:
    print(f'List có {n} phần tử (>3)')

# Trong vòng lặp đọc input đến khi rỗng:
vals = iter(['a', 'b', '', 'c'])
while (v := next(vals)) != '':
    print(v)
````

> **Cẩn thận:** dùng walrus đúng chỗ thì gọn, sai chỗ thì rối. Nếu nghi ngờ → tách dòng.

## 7. Giải thuật cơ bản _( Basic Algorithms )_


Các giải thuật kinh điển. Học để **hiểu cách hoạt động** — trong thực tế Python đã có sẵn `sorted()`, `max()`, `in`, ... rất tối ưu.

### 7.1. Linear search — tìm giá trị trong list không sắp xếp?

Duyệt từng phần tử, dừng khi gặp. Độ phức tạp **O(n)**.

Trong Python thực tế: `if x in ds` đã làm việc này, `ds.index(x)` trả về vị trí đầu tiên (raise nếu không có).

---

### Vì sao gọi là 'linear'?

Số bước tìm kiếm **tỉ lệ tuyến tính (linear)** với `n` (kích thước list):
- List 100 phần tử → tệ nhất 100 bước.
- List 1000 → 1000 bước.
- List 1tr → 1tr bước.

Trên đồ thị, đây là đường thẳng — gọi linear.

### Best / Average / Worst case

| Tình huống | Số bước |
|------------|---------|
| Phần tử ở **đầu list** (best case) | 1 |
| Phần tử ở giữa (average) | n/2 |
| Phần tử ở cuối hoặc **không tồn tại** (worst) | n |

Big-O nói về **worst case** → linear search là O(n).

### `in` vs `index` vs vòng for thủ công

```python
ds = [3, 7, 1, 9, 5, 2]

# 1. Chỉ cần biết có/không?
if 9 in ds: ...

# 2. Cần vị trí?
ds.index(9)         # 3 — vị trí đầu tiên
# ds.index(100)     # ValueError nếu không có

# 3. An toàn (trả -1 nếu không có):
next((i for i, x in enumerate(ds) if x == 9), -1)
```

### Khi linear search đủ?

- List nhỏ (vài chục → vài nghìn phần tử) → nhanh đủ.
- Tìm 1 lần (không lặp lại) → thiết lập index/tree mất công.
- Phần tử **không hashable** → không thể dùng set.

### Khi linear search là **bottleneck**?

Nếu bạn check `x in ds` **trong vòng lặp** với list lớn:
```python
# O(n²) — chậm chết!
for x in big_data:
    if x in another_big_list:
        ...

# Sửa: convert thành set (O(1) lookup) → tổng O(n)
lookup = set(another_big_list)
for x in big_data:
    if x in lookup:
        ...
```


````python
def linear_search(ds, target):
    for i, x in enumerate(ds):
        if x == target:
            return i
    return -1

ds = [3, 7, 1, 9, 5, 2]
print(linear_search(ds, 9))    # 3
print(linear_search(ds, 100))  # -1
````

### 7.2. Binary search — tìm giá trị trong list ĐÃ SẮP XẾP?

Chia đôi mỗi bước. Độ phức tạp **O(log n)** — siêu nhanh cho list lớn.

**Yêu cầu**: list phải đã sắp xếp.  
Python có sẵn module `bisect` — dùng cho production.

![Binary search divide-and-conquer](../assets/images/appendix/binary_search.png)

---

### Trực giác: chơi 'đoán số'

Bạn nghĩ ra số 1-100, mình đoán theo cách:
1. Đoán 50 — cao quá → mình biết số nằm trong 1-49.
2. Đoán 25 — thấp quá → 26-49.
3. Đoán 37 — ...

Mỗi lần **chia đôi không gian tìm kiếm**. 100 → 50 → 25 → 12 → 6 → 3 → 1 → tìm ra. Chỉ ~7 bước cho 100 số. Đó chính là binary search.

### Vì sao là **O(log n)**?

Mỗi bước chia đôi → sau `k` bước, không gian còn `n/2^k` phần tử. Để còn 1 phần tử: `n/2^k = 1 ⟹ k = log₂(n)`.

| n | log₂(n) |
|---|---------|
| 1.000 | ~10 |
| 1.000.000 | ~20 |
| 1.000.000.000 | ~30 |

**1 tỷ phần tử chỉ cần 30 bước**. Đây là sức mạnh của log.

### Phân tích invariant

```python
def binary_search(ds, target):
    lo, hi = 0, len(ds) - 1   # khoảng còn phải xét: [lo, hi]
    while lo <= hi:
        mid = (lo + hi) // 2
        if ds[mid] == target:
            return mid
        elif ds[mid] < target:
            lo = mid + 1      # target ở nửa phải
        else:
            hi = mid - 1      # target ở nửa trái
    return -1                  # không tìm thấy
```

**Invariant** = điều luôn đúng trong vòng lặp:
- Trong `[lo, hi]`: target có thể nằm ở đây.
- Ngoài `[lo, hi]`: target chắc chắn không có.

Mỗi bước thu hẹp khoảng — vòng lặp dừng khi `lo > hi` (khoảng rỗng).

### Bẫy 1: `mid = (lo + hi) // 2` có thể tràn số trong C/Java

Trong Python int **không tràn**, nên `(lo + hi) // 2` an toàn. Nhưng trong C/Java, có thể overflow → dùng `lo + (hi - lo) // 2`.

### Bẫy 2: list **chưa sort** → kết quả vô nghĩa

Binary search **giả định list đã sort**. Nếu chưa, kết quả không phải lỗi mà là sai logic — dễ bug.

**Phải sort trước** (mất O(n log n)) → chỉ đáng nếu **tìm nhiều lần** trên cùng list. Tìm 1-2 lần → linear search nhanh hơn (vì sort tốn O(n log n) > O(n)).

### Module `bisect` — phiên bản C-level

```python
import bisect
bisect.bisect_left(ds, x)     # vị trí ĐẦU TIÊN ds[i] >= x
bisect.bisect_right(ds, x)    # vị trí ĐẦU TIÊN ds[i] > x
bisect.insort(ds, x)          # chèn x giữ list sort
```

Dùng cho production — nhanh hơn cài đặt thủ công nhờ C.

### Ứng dụng ngoài tìm kiếm thuần

Binary search còn dùng để giải nhiều bài 'monotonic predicate':
- Tìm giá trị nhỏ nhất `x` mà `check(x)` đúng.
- Bài toán '**tìm tham số**' khi có hàm tăng/giảm đơn điệu.

Đây là kỹ thuật phỏng vấn rất phổ biến — học sâu hơn ở các khoá thi đấu (competitive programming).

````python
def binary_search(ds, target):
    lo, hi = 0, len(ds) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if ds[mid] == target:
            return mid
        elif ds[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

ds = [1, 3, 5, 7, 9, 11, 13, 15]
print(binary_search(ds, 7))     # 3
print(binary_search(ds, 4))     # -1

# Module bisect — chuẩn Python:
import bisect
i = bisect.bisect_left(ds, 7)
print('bisect:', i, ds[i] if i < len(ds) and ds[i] == 7 else 'không có')
````

### 7.3. Bubble sort — sắp xếp đơn giản nhất?

Đổi chỗ các cặp liền kề sai thứ tự, lặp lại đến khi không còn đổi nữa. **O(n²)** — chậm với list lớn.

![Bubble sort animation](../assets/images/appendix/bubble_sort.png)

---

### Vì sao gọi là 'bubble' (bong bóng)?

Sau mỗi 'lượt' (pass) qua list, **phần tử lớn nhất 'nổi lên'** vị trí cuối — giống bong bóng nổi lên mặt nước.

Cụ thể, sau lượt 1, ds[n-1] là max. Sau lượt 2, ds[n-2] là max của phần còn lại. Sau k lượt, k phần tử cuối đã đúng vị trí.

### Phân tích từng bước

```python
for i in range(n - 1):           # n-1 lượt là đủ
    swapped = False              # cờ: lượt này có swap nào không?
    for j in range(n - 1 - i):   # phần đuôi đã sort, không cần check
        if a[j] > a[j + 1]:
            a[j], a[j + 1] = a[j + 1], a[j]
            swapped = True
    if not swapped:              # đã sort xong → thoát sớm
        break
```

**Tối ưu 1** — `range(n - 1 - i)`: bỏ qua phần đuôi đã sort.  
**Tối ưu 2** — `swapped` cờ: nếu 1 lượt không swap → đã sort, thoát luôn.

### Tại sao O(n²)?

- Vòng ngoài: tối đa n-1 lần.
- Vòng trong: trung bình ~n/2 phép so sánh.
- Tổng: ~(n-1) × (n/2) ≈ n²/2 → O(n²).

Với n=1000: ~500.000 phép. Với n=10.000: 50 triệu — đã chậm.

### Best case của bubble — O(n) với tối ưu cờ

Nếu list **đã sort sẵn**, lượt đầu không có swap nào → cờ `swapped=False` → thoát ngay sau 1 lượt = O(n).

Đây là điểm **đặc biệt** của bubble: best case O(n), khác `selection sort` (luôn O(n²)).

### Stable sort?

✓ Có. 2 phần tử bằng nhau **không bao giờ swap** với nhau (do điều kiện `a[j] > a[j+1]` strict). → Thứ tự gốc của các phần tử bằng được giữ.

### Khi nào nên dùng?

**Trong thực tế: gần như không bao giờ** — Python có `sorted()` (Timsort) chạy O(n log n).

Bubble sort chỉ dùng để **học**: đơn giản, dễ visualize, dạy concept 'so sánh và đổi chỗ'.

````python
def bubble_sort(ds):
    a = ds[:]   # tạo bản sao
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True
        if not swapped:
            break
    return a

print(bubble_sort([5, 2, 8, 1, 9, 3]))
````

### 7.4. Selection sort?

Mỗi bước, **chọn phần tử nhỏ nhất** trong phần còn lại và đổi chỗ vào đầu. **O(n²)** nhưng số swap ít hơn bubble.

![Selection sort animation](../assets/images/appendix/selection_sort.png)

---

### Trực giác

1. Tìm min của a[0..n-1] → đổi vào vị trí 0.
2. Tìm min của a[1..n-1] → đổi vào vị trí 1.
3. Tìm min của a[2..n-1] → đổi vào vị trí 2.
...

Sau bước k, k phần tử đầu đã đúng vị trí và đã sort.

### So sánh với bubble

| Đặc điểm | Bubble | Selection |
|----------|--------|-----------|
| Số phép **so sánh** | n²/2 | n²/2 — bằng nhau |
| Số phép **swap** | tối đa n²/2 | tối đa n-1 |
| Best case | **O(n)** với tối ưu cờ | luôn O(n²) |
| Stable? | ✓ Có | ❌ Không (xem dưới) |

### Vì sao selection **không** stable?

Khi swap min vào vị trí đầu, phần tử bị 'thay chỗ' có thể nhảy qua các phần tử bằng nhau.

Ví dụ: `[2a, 4, 2b, 1]` (2a, 2b để phân biệt).
- Tìm min = 1 → swap vào index 0 → `[1, 4, 2b, 2a]`. 
- 2a giờ ở sau 2b → thứ tự gốc bị đảo.

Có thể fix bằng cách **shift thay vì swap** nhưng thường không làm — nếu cần stable, dùng insertion sort.

### Khi nào selection > bubble?

Selection có **ít swap nhất** trong các sort O(n²). Nếu mỗi swap **đắt** (ví dụ swap object lớn, ghi đĩa, gửi mạng), selection thắng. Khi swap rẻ (RAM thuần), khác biệt không đáng kể.

````python
def selection_sort(ds):
    a = ds[:]
    n = len(a)
    for i in range(n):
        min_i = i
        for j in range(i + 1, n):
            if a[j] < a[min_i]:
                min_i = j
        a[i], a[min_i] = a[min_i], a[i]
    return a

print(selection_sort([5, 2, 8, 1, 9, 3]))
````

### 7.5. Insertion sort?

Giống xếp bài tây trên tay — lấy từng lá và **chèn vào đúng chỗ** trong phần đã sắp. **O(n²)** nhưng nhanh với list **gần như đã sắp**.

---

### Trực giác — như xếp bài

Bạn có 1 chồng bài, đang xếp lên tay theo thứ tự tăng dần:
1. Lá đầu — đặt lên tay.
2. Lá thứ 2 — so với lá đầu, chèn trước hoặc sau.
3. Lá thứ 3 — so từ phải qua trái với các lá trên tay, chèn vào vị trí đúng.
...

Mỗi lá mới được 'insert' vào vị trí đúng → tên 'insertion'.

### Phân tích

```python
for i in range(1, len(a)):
    cur = a[i]
    j = i - 1
    while j >= 0 and a[j] > cur:
        a[j + 1] = a[j]    # dịch phần tử lớn sang phải
        j -= 1
    a[j + 1] = cur          # đặt cur vào vị trí đúng
```

**Quan trọng**: insertion **dịch** thay vì swap → mỗi lá chỉ swap 1 lần ở cuối, ít tốn hơn bubble (swap mỗi cặp).

### Best case **O(n)** — vì sao?

Nếu list đã sort, vòng `while` không chạy lần nào (`a[j] > cur` luôn False ngay từ đầu). Mỗi vòng ngoài chỉ làm 1 phép so sánh → tổng O(n).

Đây là lý do insertion **siêu nhanh với list 'gần như đã sort'** (vài phần tử lệch chỗ).

### Stable?

✓ Có — điều kiện `a[j] > cur` strict (không `>=`), 2 phần tử bằng không bị 'nhảy' qua nhau.

### Khi insertion thực sự được dùng

Mặc dù worst O(n²), insertion sort được dùng **bên trong** các thuật toán sort hiện đại (Timsort của Python, Introsort của C++) cho các đoạn nhỏ (~16-32 phần tử) — vì:
- Constant factor rất nhỏ (ít phép toán mỗi bước).
- Adaptive (tốt với data gần sort).
- Stable (giữ thứ tự ban đầu).

Khi n nhỏ, insertion **thực tế nhanh hơn** quicksort/mergesort dù big-O xấu hơn.

````python
def insertion_sort(ds):
    a = ds[:]
    for i in range(1, len(a)):
        cur = a[i]
        j = i - 1
        while j >= 0 and a[j] > cur:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = cur
    return a

print(insertion_sort([5, 2, 8, 1, 9, 3]))
````

### 7.6. Two pointers — kỹ thuật 2 con trỏ?

Đặt 2 chỉ số `i, j` (thường ở 2 đầu hoặc cùng đầu) và di chuyển có chiến lược. Hiệu quả cho:
- Kiểm tra palindrome.
- Two-sum trong list **đã sắp**.
- Trộn 2 list đã sắp.
- Đảo list tại chỗ.

![Two pointers - palindrome check](../assets/images/appendix/two_pointers.png)

---

### 2 dạng phổ biến

**Dạng 1 — opposite ends (2 đầu vào giữa):**
```
  i →           ← j
[ . . . . . . . . . ]
```
Dùng cho: palindrome, two-sum sorted, reverse, ...

**Dạng 2 — same direction (cả 2 đi cùng chiều):**
```
  slow →   fast →
[ . . . . . . . . . ]
```
Dùng cho: tìm cycle, dedupe in-place, sliding window.

### Vì sao two pointers **giảm O(n²) → O(n)**?

Bài toán **Two-Sum** trong list đã sort: tìm 2 số cộng = target.

**Cách ngây thơ — O(n²):**
```python
for i in range(len(ds)):
    for j in range(i + 1, len(ds)):
        if ds[i] + ds[j] == target:
            return (i, j)
```

**Two pointers — O(n):**
```python
i, j = 0, len(ds) - 1
while i < j:
    s = ds[i] + ds[j]
    if s == target: return (i, j)
    if s < target: i += 1     # cần tổng lớn hơn → tăng i
    else: j -= 1               # cần tổng nhỏ hơn → giảm j
```

**Cốt lõi**: tận dụng tính đã sort — nếu tổng quá nhỏ, KHÔNG cần thử ds[i] với mọi j sau (vì chúng đều ≥ ds[j+1] = ds[hiện tại]). Chỉ cần `i += 1` thôi.

Mỗi vòng lặp, **một trong 2 con trỏ tiến lên 1** → tối đa n bước → O(n).

### Invariant của two pointers (opposite ends)

- Cặp đáp án nếu có, nằm trong `[i, j]` (chưa loại trừ).
- Mọi cặp ngoài `[i, j]` đã được chứng minh không phải đáp án.

Khi `i >= j`: đã xét hết → kết luận không có.

### Bẫy: list **chưa sort**

Two pointers **dựa vào tính sort**. Với list chưa sort, kết luận `s < target → i += 1` sai. Phải sort trước (mất O(n log n) — vẫn nhanh hơn O(n²) ngây thơ với n lớn).

````python
# 1. Palindrome bằng 2 con trỏ:
def palindrome(s):
    i, j = 0, len(s) - 1
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True

print(palindrome('racecar'))
print(palindrome('python'))

# 2. Two-sum trong list đã sắp:
def two_sum_sorted(ds, target):
    i, j = 0, len(ds) - 1
    while i < j:
        s = ds[i] + ds[j]
        if s == target:
            return (i, j)
        if s < target:
            i += 1
        else:
            j -= 1
    return None

print(two_sum_sorted([1, 2, 4, 7, 11, 15], 13))   # (2, 4)
````

### 7.7. Sliding window — cửa sổ trượt?

Duy trì một 'cửa sổ' (đoạn liên tiếp) và di chuyển nó. Hiệu quả cho 'tổng/tích/độ dài lớn nhất của đoạn dài k' kiểu O(n) thay vì O(n·k).

![Sliding window](../assets/images/appendix/sliding_window.png)

---

### Trực giác — tận dụng overlap

Bài toán: tìm tổng lớn nhất của 3 phần tử liên tiếp trong `[2, 1, 5, 1, 3, 2]`.

**Cách ngây thơ — O(n·k):**
```python
best = 0
for i in range(len(ds) - k + 1):
    s = sum(ds[i:i+k])     # tính lại tổng MỖI lần
    best = max(best, s)
```

Mỗi vòng tính lại `sum` qua k phần tử → tổng O(n·k). Với n=10⁶, k=1000 → **10⁹ phép toán**.

**Sliding window — O(n):**
```python
cur = sum(ds[:k])              # tính tổng cửa sổ đầu
best = cur
for i in range(k, len(ds)):
    cur += ds[i] - ds[i-k]     # vào 1 phần tử, ra 1 phần tử
    best = max(best, cur)
```

**Cốt lõi**: 2 cửa sổ liên tiếp **chia sẻ k-1 phần tử**. Thay vì tính lại từ đầu, chỉ cần cộng phần tử mới vào và trừ phần tử cũ ra → **O(1) cập nhật** → tổng O(n).

### So sánh

| n | k | Ngây thơ O(n·k) | Sliding O(n) | Tăng tốc |
|---|---|------------------|---------------|----------|
| 10⁴ | 100 | 10⁶ | 10⁴ | 100× |
| 10⁶ | 1000 | 10⁹ | 10⁶ | 1000× |
| 10⁶ | 10⁵ | 10¹¹ | 10⁶ | 10⁵× |

### 2 loại sliding window

**Loại 1 — Cửa sổ cố định** (k cho trước):
```python
for i in range(k, n):
    cur += a[i] - a[i-k]
```

**Loại 2 — Cửa sổ co giãn** (kích thước thay đổi theo điều kiện):
```python
left = 0
for right in range(n):
    # mở rộng cửa sổ về phải
    while not condition_ok(window):
        # thu hẹp từ trái cho đến khi OK
        left += 1
    # cập nhật kết quả
```
Loại 2 dùng cho 'đoạn dài nhất / ngắn nhất thoả điều kiện'.

### Bẫy: chỉ áp dụng được với **thao tác cập nhật được O(1)**

Sliding window dựa vào việc **thêm/bớt 1 phần tử** vào/ra cửa sổ ở O(1):
- ✓ Tổng (cộng/trừ).
- ✓ Số phần tử dương (đếm).
- ✓ Min/max nếu dùng `deque` (monotonic queue).
- ❌ Trung vị — phức tạp hơn (cần 2 heap).
- ❌ Mode — phức tạp.

````python
# Tổng lớn nhất của k phần tử liên tiếp:
def max_sum_k(ds, k):
    if len(ds) < k:
        return None
    cur = sum(ds[:k])
    best = cur
    for i in range(k, len(ds)):
        cur += ds[i] - ds[i - k]   # vào 1 phần tử, ra 1 phần tử
        best = max(best, cur)
    return best

print(max_sum_k([2, 1, 5, 1, 3, 2], 3))   # 9 = 5+1+3
````

### 7.8. Đệ quy — giai thừa và Fibonacci?

**Đệ quy (recursion)** là hàm tự gọi chính nó.  
Mỗi đệ quy phải có **base case** (điểm dừng), nếu không sẽ tràn stack.

Lưu ý: `fib` đệ quy thuần là **O(2^n)** — siêu chậm. Cải thiện bằng **memoization** với `@lru_cache`.

![Cây đệ quy fib(5)](../assets/images/appendix/recursion_tree.png)

---

### 2 phần bắt buộc của đệ quy

1. **Base case** — điều kiện dừng, trả kết quả trực tiếp không gọi đệ quy.
2. **Recursive case** — chia bài toán thành phiên bản nhỏ hơn của chính nó.

```python
def giai_thua(n):
    if n <= 1:               # ◄── base case
        return 1
    return n * giai_thua(n - 1)   # ◄── recursive case (n nhỏ hơn)
```

**Quy tắc**: mỗi recursive call phải đến **gần hơn** với base case. Nếu không, đệ quy mãi mãi → `RecursionError: maximum recursion depth exceeded`.

### Vì sao Fibonacci đệ quy thuần là O(2^n)?

```
fib(5)
├── fib(4)
│   ├── fib(3)
│   │   ├── fib(2)            ◄── tính từ đầu
│   │   └── fib(1)
│   └── fib(2)                ◄── tính LẠI
└── fib(3)                    ◄── tính LẠI
    ├── fib(2)                ◄── tính LẠI
    └── fib(1)
```

Mỗi gọi tách thành 2 gọi nhỏ, sâu n cấp → 2^n phép tính. Với n=40: **~1.1 tỷ phép**.

Vấn đề: **tính lại quá nhiều lần** cùng 1 giá trị (`fib(2)` được tính nhiều lần).

### Memoization với `@lru_cache` — fix O(n)

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n < 2: return n
    return fib(n-1) + fib(n-2)
```

**Cốt lõi**: `lru_cache` **lưu kết quả** mỗi `fib(k)` — lần sau gọi `fib(k)` trả ngay từ cache thay vì tính lại.

→ Mỗi `fib(0)..fib(n)` chỉ tính 1 lần → **O(n)** thay vì O(2^n).

### Đệ quy vs vòng lặp — khi nào dùng cái nào?

**Đệ quy** đẹp khi bài toán **tự nhiên là 'cây'** hoặc 'chia để trị':
- Duyệt cây / đồ thị (DFS).
- Quicksort, mergesort.
- Bài toán toán học có công thức đệ quy (Fib, giai thừa).

**Vòng lặp** thắng khi:
- Đệ quy 'tail-recursive' (chỉ 1 nhánh) — Python KHÔNG tối ưu tail call → vòng lặp nhanh hơn và không tràn stack.
- Cần xử lý dữ liệu lớn (Python giới hạn ~1000 cấp đệ quy mặc định).

Vd: Fibonacci nên dùng vòng lặp với 2 biến `a, b`:
```python
def fib_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```
→ O(n) thời gian, **O(1) bộ nhớ**, không tràn stack.

### Cảnh báo: stack depth

Python mặc định cho phép ~1000 lệnh đệ quy lồng. Vượt → `RecursionError`. Tăng giới hạn bằng `sys.setrecursionlimit(10000)` — nhưng tốt hơn là viết lại bằng vòng lặp.

````python
def giai_thua(n):
    if n <= 1:           # base case
        return 1
    return n * giai_thua(n - 1)

print(giai_thua(5))   # 120

# Fibonacci — đệ quy thuần (chậm với n lớn):
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print([fib(i) for i in range(10)])

# Fibonacci — memoized:
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_fast(n):
    if n < 2:
        return n
    return fib_fast(n - 1) + fib_fast(n - 2)

print(fib_fast(50))   # tức thì

# Hoặc lặp — nhanh và không tốn stack:
def fib_iter(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(fib_iter(50))
````

> Quy tắc: nếu giải được bằng **vòng lặp**, ưu tiên vòng lặp. Đệ quy đẹp về mặt toán nhưng tốn stack.

### 7.9. Big-O notation — đo độ phức tạp?

**Big-O** mô tả tốc độ tăng thời gian/bộ nhớ theo `n` (kích thước input).

| Big-O | Tên | Ví dụ |
|-------|-----|-------|
| O(1) | hằng số | truy cập `lst[i]`, `dict[k]` |
| O(log n) | logarit | binary search |
| O(n) | tuyến tính | duyệt 1 lần qua list |
| O(n log n) | tuyến tính-log | merge sort, `sorted()` |
| O(n²) | bậc 2 | bubble/selection/insertion sort |
| O(2^n) | mũ | fibonacci đệ quy thuần |

Quy tắc thực dụng:
- n ≤ 100 → O(n³) cũng ổn.
- n ≤ 10⁴ → O(n²) ổn.
- n ≤ 10⁶ → O(n) hoặc O(n log n).
- n ≥ 10⁹ → cần O(log n) hoặc O(1).

![Big-O growth curves](../assets/images/appendix/big_o.png)

---

### Ý nghĩa thực sự của Big-O

Big-O **không** đo thời gian thực (giây), mà đo **tốc độ tăng** khi n → ∞.

Vd: 2 hàm `f(n) = 100n` và `g(n) = n² / 1000`. Big-O nói:
- `f` là O(n).
- `g` là O(n²).
- → **`f` luôn thắng `g` khi n đủ lớn** (n > 100.000), dù với n nhỏ `g` nhanh hơn.

Vì vậy Big-O **bỏ qua hằng số** và **chỉ giữ thành phần lớn nhất**:
- `100n + 50` → O(n).
- `n² + 1000n` → O(n²).
- `2^n + n^100` → O(2^n) (mũ luôn thắng đa thức khi n đủ lớn).

### Trực giác về 'bao nhiêu là đủ nhanh'?

Giả sử máy chạy ~10⁸ phép/giây (gần đúng cho Python). Để chạy trong 1 giây:

| Big-O | n tối đa cho 1 giây |
|-------|---------------------|
| O(1) | bất kỳ |
| O(log n) | 10²⁴ |
| O(n) | 10⁸ |
| O(n log n) | ~10⁶ |
| O(n²) | ~10⁴ |
| O(n³) | ~500 |
| O(2^n) | ~25 |
| O(n!) | ~10 |

Đây là tại sao Fibonacci đệ quy thuần (O(2^n)) chết với n=50 — nhưng O(n) chạy n=10⁸ vẫn ổn.

### Big-O của các thao tác Python phổ biến

| Thao tác | Big-O |
|----------|-------|
| `lst[i]`, `lst[i] = x` | O(1) |
| `lst.append(x)`, `lst.pop()` | O(1) **amortized** |
| `lst.insert(0, x)`, `lst.pop(0)` | O(n) — **shift cả list** |
| `x in lst` | O(n) |
| `x in set / dict` | O(1) trung bình |
| `lst.sort()` | O(n log n) |
| `sum(lst)`, `min(lst)` | O(n) |
| `s + s` (string) | O(n) — tạo mới |
| `''.join(lst)` | O(n) |

Hiểu các giá trị này giúp **tránh bug performance** — vd dùng `lst.pop(0)` nhiều = O(n²) ngấm ngầm.

### Space complexity

Cùng kiểu Big-O nhưng đo **bộ nhớ phụ**:
- O(1) — chỉ dùng vài biến (vòng lặp đảo list tại chỗ).
- O(n) — tạo list mới có cùng kích thước (`s[::-1]`).
- O(log n) — đệ quy có depth log n (binary search).

Khi RAM hạn chế, space complexity quan trọng không kém time.

````python
import time

def benchmark(fn, *args):
    t0 = time.perf_counter()
    fn(*args)
    return time.perf_counter() - t0

n = 50_000
ds = list(range(n))
ds_set = set(ds)

# O(n) — list `in`
print(f'list in:  {benchmark(lambda: (n - 1) in ds):.5f}s')

# O(1) — set `in`
print(f'set  in:  {benchmark(lambda: (n - 1) in ds_set):.5f}s')
````

### 8.2. Làm tròn số — `round` vs format spec?

- `round(x, n)` trả về số đã làm tròn (kiểu `float`).
- `f'{x:.{n}f}'` trả về **chuỗi** đã làm tròn để in.

Lưu ý: Python dùng **banker's rounding** (làm tròn về số chẵn) cho `round`. `round(0.5)` = 0, không phải 1!

---

### Banker's rounding là gì? Vì sao kỳ lạ?

Còn gọi là **round half to even** — nửa lên hay xuống tuỳ kết quả là chẵn:

```python
round(0.5)    # 0  — về 0 (chẵn)
round(1.5)    # 2  — về 2 (chẵn)
round(2.5)    # 2  — về 2 (chẵn)
round(3.5)    # 4  — về 4 (chẵn)
```

**Vì sao Python chọn cách này?** Vì bias = 0:
- 'Round half up' truyền thống (0.5 → 1, 1.5 → 2): **luôn lên** → bias dương khi tính trung bình nhiều lần.
- Banker's rounding: 50% lên / 50% xuống → tính tổng/trung bình của bộ dữ liệu sẽ **chính xác hơn về tổng thể**.

Đây là tiêu chuẩn IEEE 754 — được dùng trong tài chính, banking (do đó tên 'banker's').

### Khi nào nó **không** xảy ra?

Banker's rounding chỉ áp dụng khi số **chính xác bằng 0.5**. Với hầu hết các số thực, có chút sai số float, nên thường không gặp:
```python
round(0.1 + 0.1 + 0.1, 1)   # 0.3 — không phải case nửa
round(2.675, 2)              # 2.67  — bất ngờ (do 2.675 thực ra là 2.6749999...)
```

### Khi cần 'round half up' (toán học truyền thống)

Tự cài đặt:
```python
import math
def round_half_up(x, n=0):
    factor = 10 ** n
    return math.floor(x * factor + 0.5) / factor
```

Hoặc dùng `decimal` module với context chính xác:
```python
from decimal import Decimal, ROUND_HALF_UP
Decimal('0.5').quantize(Decimal('1'), rounding=ROUND_HALF_UP)   # Decimal('1')
```

### `round` vs format spec — khác biệt then chốt

- `round(x, n)` ⇒ trả về **số float**, có thể tính tiếp.
- `f'{x:.2f}'` ⇒ trả về **chuỗi**, chỉ để in/lưu/hiển thị.

**Bẫy hay gặp**:
```python
x = round(3.14159, 2)
y = x * 100        # ✓ OK — x là float

z = f'{3.14159:.2f}'
y = z * 100        # ❌ '3.143.143.14...' — z là chuỗi, * lặp chuỗi!
```

### Hai dùng chung kiểu rounding?

Có. Cả `round` và format spec đều dùng banker's rounding:
```python
round(0.5)             # 0
f'{0.5:.0f}'           # '0'
round(1.5)             # 2
f'{1.5:.0f}'           # '2'
```


````python
x = 3.14159
print(round(x, 2))         # 3.14 — float
print(f'{x:.2f}')          # '3.14' — string

# Banker's rounding bất ngờ:
print(round(0.5))          # 0
print(round(1.5))          # 2
print(round(2.5))          # 2

# Để 'làm tròn nửa lên' kiểu toán học:
import math
def round_half_up(x):
    return math.floor(x + 0.5)

print(round_half_up(2.5))  # 3
````

### 8.4. In progress bar / dòng tiến độ trong vòng lặp?

Cách đơn giản: in cùng một dòng với `\r` (carriage return) và `end=''` để không xuống dòng.

Cách 'chuẩn': dùng thư viện `tqdm` (cần `pip install tqdm`).

---

### `\r` (carriage return) là gì?

Trong văn bản, có 2 ký tự đặc biệt:
- `\n` (line feed) — **xuống dòng** (đến dòng tiếp theo).
- `\r` (carriage return) — **về đầu dòng hiện tại** (cursor về cột 0).

Khi in `'\rhello'`, cursor về đầu dòng rồi in 'hello' đè lên nội dung cũ. Đây là cơ chế của progress bar trong terminal.

**Lịch sử**: tên này đến từ **máy đánh chữ** — 'carriage' là khay giấy chạy ngang, 'return' là kéo nó về đầu hàng. `\r\n` (Windows) = về đầu hàng + xuống hàng.

### Vì sao cần `flush=True`?

`print` mặc định **buffer** output — chỉ ghi ra terminal khi:
- Buffer đầy.
- Có `\n` (xuống dòng).
- Chương trình kết thúc.

Trong vòng lặp progress bar, không có `\n` (vì dùng `end=''`) → buffer **đợi** mãi → người dùng không thấy gì cả cho đến khi xong! `flush=True` ép Python ghi ngay.

### Phân tích progress bar tự cài

```python
for i in range(n + 1):
    pct = i / n                                          # tỉ lệ 0..1
    bar = '█' * int(pct * 30) + '·' * (30 - int(pct * 30))   # 30 ký tự
    print(f'\r[{bar}] {pct * 100:5.1f}%', end='', flush=True)
    time.sleep(0.05)
print()   # xuống dòng cuối
```

- `'█' * int(pct * 30)` — tạo phần 'đầy' của bar.
- `'·' * (30 - ...)` — phần 'rỗng'.
- `\r` ở đầu — về đầu dòng.
- `end=''` — không xuống dòng (giữ nguyên dòng để đè).
- `flush=True` — buộc in ngay.
- `print()` cuối — xuống dòng để output sau không đè lên bar.

### `tqdm` — chuẩn công nghiệp

```python
from tqdm import tqdm
for x in tqdm(big_list, desc='Đang xử lý'):
    process(x)
```

Tự xử lý:
- Tốc độ (items/sec).
- Thời gian đã trôi và **dự đoán còn lại (ETA)**.
- Cập nhật bar nhanh nhưng không spam terminal.
- Tự xuống dòng đúng lúc.
