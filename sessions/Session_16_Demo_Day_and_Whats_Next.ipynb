## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Lịch demo (60 phút)

Mỗi học sinh có **3 phút** demo + **1 phút** hỏi đáp.  
Lớp ~12 học sinh ⇒ ~48 phút. Còn lại cho transition + reflexion.

**Cấu trúc demo 3 phút:**
1. **Pitch** (30s) — 1 câu giới thiệu + cảm hứng.
2. **Chạy thử** (90s) — 1 use case bình thường + 1 edge case.
3. **Code highlight** (60s) — chỉ 1 hàm bạn thích nhất / 1 đoạn xử lý lỗi.


### 2.2. Bảng feedback (peer review)

Cho mỗi bạn demo, dùng mẫu sau (in trên giấy hoặc dán ở bảng ảo):

```
Tên bạn demo:        ____
Tên dự án:           ____
1 điều ấn tượng:     ____
1 ý tưởng cải thiện: ____
1 câu hỏi:           ____
Điểm yêu thích /10:  ____
```

Cuối buổi, chuyển feedback cho bạn được review.


### 2.3. Tổng kết hành trình 16 buổi

Bạn đã học được:

- **Buổi 1-3**: Cài đặt, biến, kiểu, chuỗi, input/output.
- **Buổi 4-6**: Điều kiện, vòng lặp while/for, range.
- **Buổi 7-9**: List, tuple, set, dict, dữ liệu lồng, list comprehension.
- **Buổi 10-11**: Hàm, scope, modules (math, random, datetime).
- **Buổi 12-13**: File I/O, CSV, pandas teaser, try/except, debugging.
- **Buổi 14-16**: Capstone — bạn vừa **xây và demo** dự án Python đầu đời.

Đó là **nền tảng vững chắc** để bước vào bất kỳ hướng đi nào.


### 2.4. Hành trình tiếp theo / What's next?

**Lập trình hướng đối tượng (OOP):**
- `class`, `__init__`, `self`, kế thừa.
- Khoá học gợi ý: 'Python OOP' ở Real Python.

**Web với Flask hoặc FastAPI:**
- Biến script Python thành **website** mà bạn bè vào dùng được.
- Bắt đầu với Flask: 1 file, 30 dòng code → có web.

**Data / AI:**
- Đào sâu `pandas`, `numpy`, `matplotlib`.
- Học `scikit-learn` cho ML cổ điển; sau đó là deep learning với PyTorch.
- Kaggle Learn (miễn phí) là nơi tốt để bắt đầu.

**Automation:**
- Cuốn 'Automate the Boring Stuff with Python' (miễn phí online).
- Tự động hoá: gửi email, đổi tên file hàng loạt, scrape web, gửi tin nhắn Telegram.

**Cuộc thi / cộng đồng:**
- HSG Tin học thành phố/quốc gia (THPT).
- Tham gia hackathon (HackerEarth, MLH).
- Tạo tài khoản GitHub — public dự án capstone.


### 2.5. Lời cuối

**Lập trình giống học một ngôn ngữ — bạn không cần nhớ hết từ vựng, chỉ cần biết tra từ điển và viết câu mỗi ngày.**

4. **Hoàn thành** thay vì hoàn hảo — dự án hoàn thành tốt hơn dự án 'đang dở'.
