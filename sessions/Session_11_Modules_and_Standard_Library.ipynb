## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Module là gì?

**Module** = một file `.py` chứa nhiều hàm/biến có thể tái sử dụng.

Python có sẵn rất nhiều module trong **thư viện chuẩn (standard library)** — không cần cài, chỉ cần `import`.

Ngoài ra có hàng trăm ngàn module bên ngoài (third-party): `numpy`, `pandas`, `requests`, ...


### 2.2. Hai cách `import`

**Cách 1:** import cả module — gọi qua `module.tên`.
```python
import math
math.sqrt(2)
```

**Cách 2:** import tên cụ thể — dùng trực tiếp.
```python
from math import sqrt, pi
sqrt(2)
```

**Cách 3 (đặt bí danh):**
```python
import numpy as np
import pandas as pd
```

**Tránh** `from module import *` (rối, dễ trùng tên).


### 2.3. Module `math`

| Hàm / hằng số | Tác dụng |
|---------------|----------|
| `math.pi`, `math.e` | hằng số nổi tiếng |
| `math.sqrt(x)` | căn bậc 2 |
| `math.pow(a, b)` | a luỹ thừa b (giống `a**b`) |
| `math.ceil(x)` | làm tròn lên |
| `math.floor(x)` | làm tròn xuống |
| `math.gcd(a, b)` | ước chung lớn nhất |
| `math.factorial(n)` | giai thừa |
| `math.log(x)`, `math.log10(x)` | logarit |
| `math.sin/cos/tan(x)` | lượng giác (radian) |


````python
import math

print(math.pi)
print(math.sqrt(16))
print(math.factorial(5))
print(math.gcd(48, 18))
print(math.ceil(3.2), math.floor(3.7))
print(round(math.sin(math.pi / 2), 4))
````

### 2.4. Module `random` — dữ liệu ngẫu nhiên

| Hàm | Tác dụng |
|-----|----------|
| `random.random()` | float trong `[0, 1)` |
| `random.randint(a, b)` | số nguyên trong `[a, b]` (cả 2 đầu) |
| `random.uniform(a, b)` | float trong `[a, b]` |
| `random.choice(seq)` | chọn ngẫu nhiên 1 phần tử |
| `random.sample(seq, k)` | lấy k phần tử **không trùng** |
| `random.shuffle(lst)` | xáo trộn list (sửa list gốc) |
| `random.seed(n)` | cố định mẫu ngẫu nhiên (debug) |


````python
import random

random.seed(42)   # để chạy lại cho ra cùng kết quả
print(random.random())
print(random.randint(1, 100))
print(random.choice(['kéo', 'búa', 'bao']))
print(random.sample(range(1, 50), 6))   # vé số 6 số
lst = [1, 2, 3, 4, 5]
random.shuffle(lst)
print(lst)
````

### 2.5. Module `datetime` — ngày giờ


````python
from datetime import datetime, date, timedelta

now = datetime.now()
print('Bây giờ:', now)
print('Năm:', now.year, 'Tháng:', now.month, 'Ngày:', now.day)

sinh_nhat = date(2010, 5, 22)
today = date.today()
tuoi_ngay = (today - sinh_nhat).days
print(f'Bạn sinh ngày {sinh_nhat}, đã sống {tuoi_ngay} ngày.')

tuong_lai = today + timedelta(days=100)
print('100 ngày nữa là:', tuong_lai)
````

### 2.6. Module `statistics`

Dùng nhanh khi không có pandas/numpy.


````python
import statistics as st

diem = [8, 5, 9, 4, 7, 6.5, 8.5, 7, 9]
print('TB:', st.mean(diem))
print('Trung vị:', st.median(diem))
print('Mode:', st.mode(diem))
print('Độ lệch chuẩn:', round(st.stdev(diem), 2))
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- Module = file `.py` chứa hàm/biến — `import` để dùng.
- `math` cho toán học, `random` cho ngẫu nhiên, `datetime` cho ngày giờ, `statistics` cho thống kê.
- Có thể `import x as ten_ngan` để gọn.
- Tự viết module bằng cách tạo file `.py` cùng thư mục.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `module` | module | file .py có thể import |
| `import` | import | nạp module vào notebook |
| `thư viện chuẩn` | standard library | module có sẵn của Python |
| `bí danh` | alias | import x as y — đổi tên cho gọn |
| `hạt giống` | seed | cố định ngẫu nhiên để debug |
