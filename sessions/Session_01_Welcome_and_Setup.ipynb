### 2.1. Jupyter Notebook là gì?

Bạn đang nhìn vào một **Jupyter Notebook** — một loại tài liệu đặc biệt cho phép xen kẽ:

- **Markdown cell**: ô chữ (giống Word), giải thích, hình ảnh, công thức.
- **Code cell**: ô code (mã) — chạy được. Khi bạn nhấn `Shift + Enter`, máy sẽ thực thi nó.

| Phím | Tác dụng |
|------|---------|
| `Shift + Enter` | Chạy cell, sang cell sau |
| `Ctrl + Enter` | Chạy cell, **không** sang cell sau |
| `Alt + Enter` | Chạy cell, **chèn cell mới phía dưới** |
| `A` (khi cell không edit) | Chèn cell trên |
| `B` (khi cell không edit) | Chèn cell dưới |
| `M` / `Y` | Đổi cell sang Markdown / Code |
| `D D` (gõ D 2 lần) | Xoá cell |

### 2.3. Kernel & runtime state

**Kernel** là 'cái não' Python đứng sau notebook. Khi bạn chạy 1 cell, kernel **nhớ** mọi biến và hàm bạn đã định nghĩa.

Điều quan trọng cần nhớ:

- Nếu code 'tự nhiên' chạy lạ → thử restart kernel rồi chạy lại từ trên xuống.

### 2.5. Tuỳ chỉnh `print()` — `sep` và `end`

`print()` còn có 2 tham số đặc biệt:
- `sep=` — đổi dấu ngăn cách giữa các giá trị (mặc định là khoảng trắng).
- `end=` — đổi cái sẽ in ra **sau** dòng (mặc định là `\n` — xuống dòng).


````python
print('a', 'b', 'c')                  # mặc định: 'a b c'
print('a', 'b', 'c', sep='-')         # 'a-b-c'
print('a', 'b', 'c', sep='')          # 'abc'
print('không xuống dòng', end=' ')
print('vẫn cùng 1 dòng')
````

### 2.6. Ký tự đặc biệt trong chuỗi (Escape characters)

Có những ký tự không gõ trực tiếp được — phải dùng dấu `\` (backslash) phía trước.

| Ký tự | Ý nghĩa |
|-------|---------|
| `\n` | Xuống dòng (newline) |
| `\t` | Tab — khoảng trắng lớn để căn cột |
| `\\` | In ra **một** dấu `\` |
| `\'` | In dấu `'` trong chuỗi bọc bằng `'` |
| `\"` | In dấu `"` trong chuỗi bọc bằng `"` |


````python
print('Dòng 1\nDòng 2\nDòng 3')               # 3 dòng từ 1 lệnh print
print('Cột1\tCột2\tCột3')                    # cách bằng tab
print('Đường dẫn: C:\\Users\\Lan')              # in được dấu \
print('Anh ấy nói: \'Xin chào!\'')             # escape dấu nháy đơn
print("Cô ấy hỏi: \"Bạn khoẻ không?\"")        # escape dấu nháy kép
````

**Đoán trước khi chạy / Predict before running:** Đoán: dòng `print('a\\nb')` sẽ in ra gì? Còn `print('a\\\\nb')` thì sao?

```python
print('a\nb')        # \n bình thường — xuống dòng
print('a\\nb')       # in chữ \ + chữ n + b
print('a\\\\nb')     # in 2 chữ \ + chữ n + b
```

_Hãy nghĩ xem kết quả là gì, **rồi** chạy ô bên dưới để kiểm tra._

### 2.8. Comment — chú thích trong code

Bất cứ thứ gì sau dấu `#` trên cùng một dòng đều là **comment (chú thích)** — máy bỏ qua, chỉ con người đọc.

Comment dùng để:
- **Giải thích** code cho chính bạn (3 tháng sau quay lại sẽ cảm ơn bạn của hiện tại).
- **Tạm vô hiệu hoá** một dòng để thử (rất phổ biến khi debug).
- **Đánh dấu TODO** cho phần chưa làm: `# TODO: ...`

**Lưu ý quan trọng:** comment giải thích **'tại sao'**, không phải **'làm gì'**. Code đã nói nó làm gì — comment chỉ cần thiết khi code không thể tự nói rõ ý đồ.


````python
# Đây là comment — máy không chạy dòng này
print('Dòng này có chạy.')  # comment cũng có thể đứng sau code
# print('Dòng này KHÔNG chạy vì bị comment ra.')

# TODO: thêm tính năng in tên người dùng

# Comment xấu — chỉ lặp lại điều code đã nói:
# In từ 'Hello'
print('Hello')

# Comment tốt — giải thích lý do:
# Dùng print 2 lần để dễ thấy trong terminal log
print('Hello')
````

### 2.9. Lỗi (errors) — bạn của lập trình viên

Khi viết sai cú pháp, Python sẽ báo **lỗi (error)**. Đừng sợ — đọc kỹ thông báo, nó luôn chỉ chỗ sai.

|-----|---------|
| `SyntaxError` | Viết sai cú pháp (thiếu `)`, lệch ngoặc, ...) |
| `NameError` | Dùng tên biến/hàm chưa được định nghĩa |
| `IndentationError` | Thụt lề sai |

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- Python là ngôn ngữ lập trình dễ học, mạnh mẽ, được dùng khắp nơi.
- Jupyter Notebook xen kẽ ô **Markdown** (chữ) và ô **Code** (chạy được).
- `print()` in một thứ gì đó ra màn hình.
- `#` bắt đầu một **comment** — máy bỏ qua.
- Lỗi (error) không đáng sợ — đọc kỹ dòng cuối để biết lỗi gì.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `khoá học` | Khoá học / course | 8 tuần, 16 buổi, mỗi buổi 90 phút |
| `sổ tay điện tử` | notebook | file `.ipynb` xen kẽ chữ và code |
| `ô lệnh` | cell | đơn vị nhỏ nhất của notebook, 2 loại: Markdown & Code |
| `định dạng văn bản` | Markdown | ngôn ngữ định dạng chữ đơn giản |
| `mã (lệnh)` | code | phần Python sẽ chạy |
| `nhân` | kernel | tiến trình Python đang chạy trong notebook |
| `in ra` | print | hiển thị nội dung ra màn hình |
| `chú thích` | comment | bắt đầu bằng `#`, máy bỏ qua |
| `lỗi` | error | thông báo khi code sai cú pháp hoặc logic |
| `cú pháp` | syntax | luật viết câu lệnh đúng |
