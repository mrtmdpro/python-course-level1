## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. List là gì?

**List (danh sách)** là cấu trúc lưu **nhiều giá trị có thứ tự** trong một biến duy nhất.

Cú pháp: dấu ngoặc vuông `[ ]`, các phần tử cách nhau bởi `,`.


````python
diem = [8.5, 7, 9, 6.5, 10]
ten = ['An', 'Bình', 'Châu']
hon_hop = [1, 'hai', 3.0, True]
rong = []

print(diem)
print(len(diem))
````

### 2.2. Index và slice — giống chuỗi

Cách truy cập y hệt chuỗi: `lst[i]`, `lst[a:b]`, hỗ trợ index âm.


````python
diem = [8.5, 7, 9, 6.5, 10]
print(diem[0])      # 8.5
print(diem[-1])     # 10
print(diem[1:4])    # [7, 9, 6.5]
print(diem[::-1])   # đảo
````

### 2.3. List **thay đổi được** (mutable) — khác chuỗi

Bạn có thể **gán lại** từng phần tử, **thêm**, **xoá**, **sắp xếp** một list.
Chuỗi thì KHÔNG đổi được — phải tạo chuỗi mới.


````python
diem = [8.5, 7, 9]
diem[1] = 7.5     # gán lại phần tử
print(diem)
````

### 2.4. Phương thức quan trọng

| Method | Tác dụng |
|--------|----------|
| `lst.append(x)` | thêm `x` vào cuối |
| `lst.extend(other)` | nối 1 list khác vào cuối |
| `lst.insert(i, x)` | chèn `x` vào vị trí `i` |
| `lst.remove(x)` | xoá lần xuất hiện đầu tiên của `x` |
| `lst.pop()` / `lst.pop(i)` | lấy ra (và xoá) phần tử cuối / vị trí `i` |
| `lst.sort()` | sắp xếp tăng dần (sửa list gốc) |
| `lst.sort(reverse=True)` | sắp xếp giảm dần |
| `lst.reverse()` | đảo chỗ |
| `lst.count(x)` | đếm số lần `x` xuất hiện |
| `lst.index(x)` | trả về index đầu tiên của `x` |


````python
ds = [3, 1, 4, 1, 5]
ds.append(9)
ds.append(2)
print(ds)
ds.sort()
print(ds)
ds.reverse()
print(ds)
print('Số lần xuất hiện 1:', ds.count(1))
ds.remove(1)         # chỉ xoá lần đầu
print(ds)
````

### 2.5. Toán tử với list

- `+` nối 2 list: `[1,2] + [3,4]` → `[1,2,3,4]`
- `*` lặp: `[0] * 5` → `[0, 0, 0, 0, 0]`
- `in` kiểm tra: `3 in [1,2,3]` → `True`


````python
a = [1, 2, 3]
b = [4, 5]
print(a + b)
print([0] * 5)
print(2 in a)
````

### 2.6. Duyệt list bằng `for`

Hai cách phổ biến: duyệt theo **giá trị**, hoặc duyệt theo **index**.


````python
ten = ['An', 'Bình', 'Châu']

# Cách 1: duyệt giá trị
for x in ten:
    print('Chào', x)

# Cách 2: duyệt index
for i in range(len(ten)):
    print(i + 1, '-', ten[i])

# Cách 3: duyệt cả 2 với enumerate
for i, x in enumerate(ten, start=1):
    print(f'{i}. {x}')
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- List `[ ]` lưu nhiều giá trị có thứ tự, có thể đổi (mutable).
- Truy cập bằng `lst[i]`, cắt bằng `lst[a:b]`, độ dài bằng `len()`.
- Phương thức: `append, extend, insert, remove, pop, sort, reverse, count, index`.
- Toán tử: `+`, `*`, `in`. Hàm: `min, max, sum, sorted`.
- Duyệt: `for x in lst` hoặc `for i, x in enumerate(lst)`.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `danh sách` | list | cấu trúc lưu nhiều giá trị có thứ tự |
| `thay đổi được` | mutable | có thể chỉnh sửa sau khi tạo |
| `không đổi được` | immutable | ví dụ chuỗi, tuple |
| `thêm vào cuối` | append | lst.append(x) |
| `đếm + duyệt` | enumerate | for i, x in enumerate(lst) |
| `sort vs sorted` | sort vs sorted | sort sửa list gốc; sorted trả về list mới |
