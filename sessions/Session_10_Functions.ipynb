### 2.1. Tại sao cần hàm?

Hãy tưởng tượng bạn cần tính BMI cho 100 học sinh. Nếu copy code 100 lần — sai 1 chỗ là sai 100 chỗ.

**Hàm (function)** = gói một đoạn code có thể **tái sử dụng** nhiều lần với input khác nhau.

Lợi ích:
- **DRY** (Don't Repeat Yourself): viết một lần, gọi nhiều nơi.
- **Đặt tên** cho thao tác → code đọc như tiếng Anh.
- **Test** dễ hơn, **sửa lỗi** chỉ một chỗ.


### 2.2. Cú pháp `def`

```python
def tên_hàm(tham_số_1, tham_số_2):
    """Mô tả ngắn gọn (docstring)."""
    # khối lệnh
    return giá_trị
```

Sau khi `def`, hàm chỉ được **định nghĩa**; phải **gọi** mới chạy.


````python
def chao(ten):
    """Trả về một câu chào."""
    return f'Xin chào, {ten}!'

print(chao('An'))
print(chao('Bình'))
print(chao('Châu'))
````

### 2.3. `return` vs `print`

Đây là chỗ rất nhiều bạn nhầm.

- `print(x)`: **hiển thị** ra màn hình. Hàm không có `return` thì trả về `None`.
- `return x`: **trả về giá trị** cho người gọi → có thể gán vào biến hoặc dùng tiếp.


````python
def cong_in(a, b):
    print(a + b)         # chỉ in, không trả về

def cong_tra(a, b):
    return a + b         # trả về kết quả

x = cong_in(2, 3)        # in '5'
print('x =', x)          # x = None

y = cong_tra(2, 3)
print('y =', y)          # y = 5
print(cong_tra(2, 3) * 10)  # 50 — dùng kết quả tiếp
````

### 2.4. Tham số mặc định và keyword args

Cho phép tham số có **giá trị mặc định** — gọi hàm gọn hơn.


````python
def chao(ten, loi='Xin chào'):
    return f'{loi}, {ten}!'

print(chao('An'))                     # Xin chào, An!
print(chao('Bình', 'Hi'))            # Hi, Bình!
print(chao(loi='Hello', ten='An'))   # gọi bằng tên (keyword)
````

### 2.5. Trả về nhiều giá trị (qua tuple)


````python
def thong_ke(ds):
    """Trả về (nhỏ nhất, lớn nhất, trung bình) của list."""
    return min(ds), max(ds), sum(ds) / len(ds)

lo, hi, tb = thong_ke([3, 7, 1, 9, 5])
print('Min:', lo)
print('Max:', hi)
print(f'TB: {tb:.2f}')
````

### 2.6. Phạm vi biến / Scope

- Biến tạo **trong** hàm là **local** — chỉ tồn tại trong hàm đó.
- Biến ngoài hàm là **global** — mọi nơi đọc được, nhưng tránh sửa từ trong hàm.
- Nguyên tắc: **truyền dữ liệu vào qua tham số, lấy ra qua return**.


````python
x = 10        # global

def f():
    x = 99    # local — biến mới, KHÔNG sửa global
    print('Trong hàm:', x)

f()
print('Ngoài hàm:', x)
````

### 2.7. Tách bài toán lớn thành các hàm nhỏ

Một quy tắc: mỗi hàm nên làm **một việc** và làm tốt.


````python
def tinh_bmi(can_nang, chieu_cao):
    return can_nang / chieu_cao ** 2

def xep_bmi(bmi):
    if bmi < 18.5: return 'Gầy'
    if bmi < 25:  return 'Bình thường'
    if bmi < 30:  return 'Thừa cân'
    return 'Béo phì'

def in_bmi(can, cao):
    bmi = tinh_bmi(can, cao)
    print(f'BMI = {bmi:.2f} → {xep_bmi(bmi)}')

in_bmi(55, 1.65)
in_bmi(80, 1.70)
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- `def tên_hàm(tham_số):` định nghĩa hàm; gọi bằng `tên_hàm(...)`.

- Biến trong hàm là local; truyền vào qua tham số, lấy ra qua return.
- Tách bài toán lớn thành nhiều hàm nhỏ — mỗi hàm làm một việc.

| `def` | def | từ khoá định nghĩa hàm |
| `tham số / đối số` | parameter / argument | đầu vào của hàm |
