## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Vì sao cần file?

Mọi biến trong notebook **biến mất** khi tắt kernel. Để **lưu dữ liệu lâu dài** (ví dụ: điểm số, nhật ký, danh sách bạn bè) — ta cần ghi ra file.

Ngược lại, khi muốn **đọc dữ liệu** từ một nguồn (ví dụ: file CSV của giáo viên) — ta phải đọc từ file.


### 2.2. Cú pháp `with open()`

```python
with open('ten_file.txt', 'r', encoding='utf-8') as f:
    noi_dung = f.read()
```

**Tại sao `with`?** Nó **tự động đóng file** khi xong, kể cả khi có lỗi. Luôn dùng `with`.

**Chế độ mở (mode):**
| Mode | Tác dụng |
|------|----------|
| `'r'` | đọc (read) — file phải tồn tại |
| `'w'` | ghi (write) — **xoá nội dung cũ** nếu có |
| `'a'` | thêm vào cuối (append) |
| `'r+'` | đọc và ghi |

**Quan trọng:** thêm `encoding='utf-8'` để tiếng Việt không bị lỗi.


### 2.3. Ghi file văn bản


````python
with open('chao.txt', 'w', encoding='utf-8') as f:
    f.write('Xin chào!\n')
    f.write('Tôi đang học Python.\n')
    f.write('Hôm nay là buổi 12.\n')

print('Đã ghi xong file chao.txt')
````

### 2.4. Đọc file văn bản — 3 cách


````python
# Cách 1: đọc tất cả thành 1 chuỗi
with open('chao.txt', 'r', encoding='utf-8') as f:
    noi_dung = f.read()
print(noi_dung)
print('---')

# Cách 2: đọc thành list các dòng
with open('chao.txt', 'r', encoding='utf-8') as f:
    dong = f.readlines()
print(dong)
print('---')

# Cách 3 (khuyến nghị cho file lớn): duyệt từng dòng
with open('chao.txt', 'r', encoding='utf-8') as f:
    for d in f:
        print(d.rstrip())
````

### 2.5. Append — thêm vào cuối


````python
with open('chao.txt', 'a', encoding='utf-8') as f:
    f.write('Dòng được thêm vào cuối.\n')

with open('chao.txt', 'r', encoding='utf-8') as f:
    print(f.read())
````

### 2.6. CSV — Comma-Separated Values

**CSV** là định dạng văn bản đơn giản để lưu **bảng**: mỗi dòng là 1 hàng, các ô cách nhau bởi dấu phẩy `,`. Excel, Google Sheet đều xuất được CSV.

Module `csv` chuẩn của Python xử lý CSV an toàn hơn `split(',')` (vì có những ô chứa dấu phẩy bên trong — phải bọc bằng `"`).


````python
import csv

# Ghi CSV
rows = [
    ['ten', 'lop', 'diem'],
    ['An', '10A1', 8.5],
    ['Bình', '10A2', 7.0],
    ['Châu', '10A1', 9.0],
]
with open('diem.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Đọc CSV
with open('diem.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
````

````python
# DictReader — tự dùng dòng đầu làm key
import csv
with open('diem.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for hs in reader:
        print(f"{hs['ten']} ({hs['lop']}): {hs['diem']}")
````

### 2.7. Giới thiệu nhanh `pandas`

Trong các bài thực tế (capstone Track B), `pandas` là cách thuận tiện hơn nhiều để xử lý CSV — nó load thành **DataFrame** (giống bảng Excel).


````python
import pandas as pd

df = pd.read_csv('diem.csv')
print(df)
print()
print('Trung bình điểm:', df['diem'].mean())
print('Theo lớp:')
print(df.groupby('lop')['diem'].mean())
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- `with open(path, mode, encoding='utf-8')` — luôn dùng `with`.
- Mode: `r` đọc, `w` ghi (xoá cũ), `a` thêm cuối.
- `f.read()`, `f.readlines()`, hoặc `for d in f` để đọc.
- CSV: dùng module `csv` (`reader`, `DictReader`, `writer`).
- `pandas.read_csv` mạnh và gọn cho phân tích dữ liệu.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `tệp` | file | lưu dữ liệu lâu dài trên đĩa |
| `chế độ` | mode | r/w/a — cách mở file |
| `mã hoá` | encoding | utf-8 cho tiếng Việt |
| `CSV` | CSV | định dạng bảng phân cách bằng dấu phẩy |
| `DataFrame` | DataFrame | bảng dữ liệu của pandas |
| `nhóm theo` | groupby | gom hàng theo cột rồi tính |
