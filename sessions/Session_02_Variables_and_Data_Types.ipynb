## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Biến (variable) là gì?

**Biến** là một **cái tên** đại diện cho một **giá trị** (value) được lưu trong bộ nhớ.

Hãy tưởng tượng biến như **một cái hộp có dán nhãn**: nhãn là **tên biến**, trong hộp là **giá trị**.

Cú pháp gán (assignment):

```python
tên_biến = giá_trị
```


````python
tuoi = 16
ten = 'Lan'
chieu_cao = 1.65

print(ten)
print(tuoi)
print(chieu_cao)
````

### 2.2. Quy tắc đặt tên biến / Naming rules

**Được phép:**
- Bắt đầu bằng chữ cái hoặc dấu `_`
- Sau đó có thể có chữ, số, dấu `_`
- Phân biệt **chữ hoa và chữ thường**: `Tuoi` khác `tuoi`

**Không được phép:**
- Bắt đầu bằng số: `2tuoi` ❌
- Có dấu cách: `chieu cao` ❌ (dùng `chieu_cao`)
- Trùng từ khoá Python: `if`, `for`, `class`, ...

**Quy ước (convention):** dùng `snake_case` — chữ thường, ngăn cách bằng `_`.
Ví dụ tốt: `diem_toan`, `so_hoc_sinh`, `email_lien_he`.


### 2.3. Bốn kiểu dữ liệu cơ bản / Four basic types

| Kiểu | Tên đầy đủ | Ví dụ |
|------|------------|-------|
| `int` | integer (số nguyên) | `0`, `42`, `-7` |
| `float` | số thực (số thập phân) | `3.14`, `-0.5`, `1.0` |
| `str` | string (chuỗi ký tự) | `'Hello'`, `"Python"` |
| `bool` | boolean (logic) | `True`, `False` |

Dùng `type()` để xem kiểu của một giá trị.


````python
print(type(42))
print(type(3.14))
print(type('Xin chào'))
print(type(True))
````

### 2.4. Phép toán với số / Arithmetic

| Toán tử | Ý nghĩa | Ví dụ | Kết quả |
|---------|---------|-------|---------|
| `+` | cộng | `7 + 2` | `9` |
| `-` | trừ | `7 - 2` | `5` |
| `*` | nhân | `7 * 2` | `14` |
| `/` | chia (luôn ra float) | `7 / 2` | `3.5` |
| `//` | chia lấy phần nguyên | `7 // 2` | `3` |
| `%` | chia lấy dư (modulo) | `7 % 2` | `1` |
| `**` | luỹ thừa | `2 ** 10` | `1024` |


````python
a = 7
b = 2
print('Tổng:', a + b)
print('Hiệu:', a - b)
print('Tích:', a * b)
print('Thương:', a / b)
print('Chia nguyên:', a // b)
print('Chia dư:', a % b)
print('Luỹ thừa:', a ** b)
````

### 2.5. Chuyển kiểu / Type conversion

Đôi khi ta phải đổi kiểu thủ công, ví dụ khi cộng `'5'` (chuỗi) với `3` (số) — Python sẽ báo lỗi.

- `int(x)`  — đổi sang số nguyên
- `float(x)` — đổi sang số thực
- `str(x)`  — đổi sang chuỗi


````python
x = '5'
y = 3
# print(x + y)  # ❌ TypeError: cộng chuỗi với số
print(int(x) + y)        # 8 — đổi '5' thành 5 trước
print(x + str(y))        # '53' — nối hai chuỗi
````

### 2.6. Cập nhật giá trị biến / Updating variables

Bạn có thể **gán lại** một biến bất kỳ lúc nào. Python cũng có cú pháp viết tắt rất tiện:

| Viết tắt | Tương đương |
|---------|--------------|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 2` | `x = x * 2` |
| `x /= 2` | `x = x / 2` |


````python
diem = 7
diem += 2          # cộng thêm 2
print(diem)        # 9
diem *= 10
print(diem)        # 90
````
