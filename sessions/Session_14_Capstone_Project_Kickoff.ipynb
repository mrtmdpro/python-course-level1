### 2.6. Scaffold mẫu — Track C (matplotlib)


````python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import random
import math

def ve_da_giac(ax, x, y, n_canh, ban_kinh, mau, xoay=0):
    """Vẽ đa giác đều n cạnh tại (x, y) với bán kính cho trước."""
    diem = []
    for i in range(n_canh):
        goc = xoay + 2 * math.pi * i / n_canh
        diem.append((x + ban_kinh * math.cos(goc), y + ban_kinh * math.sin(goc)))
    poly = patches.Polygon(diem, facecolor=mau, edgecolor='black', linewidth=1)
    ax.add_patch(poly)

def ve_mosaic(so_hinh=30):
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_facecolor('#FAFAFA')
    
    mau_palette = ['#3B82F6', '#F59E0B', '#10B981', '#EF4444', '#A855F7', '#EC4899']
    
    for _ in range(so_hinh):
        x = random.uniform(1, 9)
        y = random.uniform(1, 9)
        n = random.randint(3, 8)
        r = random.uniform(0.3, 0.8)
        mau = random.choice(mau_palette)
        ve_da_giac(ax, x, y, n, r, mau, xoay=random.uniform(0, 2 * math.pi))
    
    plt.title('Mosaic — generative art')
    plt.show()
    # plt.savefig('my_art.png', dpi=150, bbox_inches='tight')   # bỏ comment để lưu file

ve_mosaic(30)
````
