## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Giá trị `True` / `False` — kiểu bool

Python có 2 giá trị logic: `True` (đúng) và `False` (sai). Chú ý: viết hoa chữ T và F.

Các **phép so sánh** trả về bool:

| Toán tử | Ý nghĩa | Ví dụ | Kết quả |
|---------|---------|-------|---------|
| `==` | bằng | `5 == 5` | `True` |
| `!=` | khác | `5 != 3` | `True` |
| `<` `>` | bé / lớn hơn | `3 < 5` | `True` |
| `<=` `>=` | bé / lớn hơn hoặc bằng | `5 >= 5` | `True` |

**Lưu ý:** dấu `=` là phép gán, `==` mới là so sánh. Đây là lỗi phổ biến!


````python
print(5 == 5)
print(5 == 6)
print('Python' == 'python')   # phân biệt hoa thường
print(10 >= 10)
print(3 != 4)
````

### 2.2. `if`, `elif`, `else`

Cú pháp:

```python
if điều_kiện_1:
    # khối lệnh 1
elif điều_kiện_2:
    # khối lệnh 2
else:
    # khối lệnh mặc định
```

**Indentation (thụt lề) bắt buộc:** các dòng cùng một khối phải thụt lề **cùng số khoảng trắng** (thường là 4). Đây là cách Python biết dòng nào thuộc khối nào.


````python
diem = 7.5
if diem >= 8:
    print('Giỏi')
elif diem >= 6.5:
    print('Khá')
elif diem >= 5:
    print('Trung bình')
else:
    print('Yếu')
````

### 2.3. Kết hợp điều kiện: `and`, `or`, `not`

| Toán tử | Trả về `True` khi |
|---------|-------------------|
| `A and B` | **cả** A và B đều True |
| `A or B` | **ít nhất một** True |
| `not A` | A là False |


````python
tuoi = 17
co_giay_to = True

if tuoi >= 18 and co_giay_to:
    print('Được lái xe.')
elif tuoi >= 16 and co_giay_to:
    print('Được lái xe máy < 50cc.')
else:
    print('Chưa được lái.')
````

### 2.4. `if` lồng nhau / Nested if

Khối `if` có thể nằm trong `if` khác. Hãy thụt lề sâu hơn để Python hiểu.


````python
diem = 8
hanh_kiem = 'Tốt'

if diem >= 8:
    if hanh_kiem == 'Tốt':
        print('Học sinh giỏi toàn diện.')
    else:
        print('Học giỏi nhưng cần rèn hạnh kiểm.')
else:
    print('Cần cố gắng thêm.')
````

### 2.5. Kiểm tra giá trị có nằm trong tập hợp

Toán tử `in` cho phép kiểm tra một giá trị có nằm trong chuỗi/list/tuple không.


````python
mon = 'Toán'
if mon in ('Toán', 'Lý', 'Hoá'):
    print('Môn tự nhiên')
elif mon in ('Văn', 'Sử', 'Địa'):
    print('Môn xã hội')
else:
    print('Môn khác')
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- `bool` chỉ có `True` / `False`. So sánh dùng `==`, không phải `=`.
- `if / elif / else` cho máy ra quyết định.
- Khối lệnh xác định bằng **indentation** — luôn 4 khoảng trắng.
- Kết hợp điều kiện bằng `and`, `or`, `not`. Toán tử `in` kiểm tra thuộc tập.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `luận lý` | boolean | chỉ có True / False |
| `so sánh` | comparison | ==, !=, <, >, <=, >= |
| `câu lệnh điều kiện` | if statement | if / elif / else |
| `thụt lề` | indentation | 4 khoảng trắng — Python dùng để xác định khối |
| `toán tử logic` | logical operator | and, or, not |
| `thuộc` | in | kiểm tra một giá trị có trong chuỗi/list/tuple |
