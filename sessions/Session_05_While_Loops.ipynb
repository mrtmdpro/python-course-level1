## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Vì sao cần vòng lặp?

Hãy in `Học Python` 100 lần. Nếu copy-paste 100 dòng, code sẽ rất xấu và dễ sai.
Vòng lặp (loop) cho phép **lặp lại** một khối lệnh cho đến khi điều kiện không còn đúng.

Python có 2 loại vòng lặp chính: **`while`** (lặp khi điều kiện đúng) và **`for`** (lặp qua từng phần tử). Hôm nay học `while`, buổi sau học `for`.


### 2.2. Cú pháp `while`

```python
while điều_kiện:
    # khối lệnh — lặp lại khi điều_kiện là True
```

Cách hoạt động:
1. Kiểm tra điều kiện.
2. Nếu True → chạy khối lệnh → quay lại bước 1.
3. Nếu False → thoát khỏi vòng lặp.


````python
i = 1
while i <= 5:
    print(f'Lần {i}: Học Python')
    i += 1   # nếu quên dòng này → vòng lặp vô hạn!
print('Xong.')
````

### 2.3. Vòng lặp vô hạn / Infinite loop

Nếu điều kiện luôn True, vòng lặp sẽ chạy mãi. Trong Colab/VS Code, bạn có thể nhấn **Stop / Interrupt** ở thanh công cụ để dừng.

Ví dụ điển hình của bug:
```python
i = 1
while i <= 5:
    print(i)
    # quên i += 1 — vòng lặp vô hạn!
```


### 2.4. `break` — thoát ngay

Khi gặp `break`, Python **thoát hẳn** khỏi vòng lặp gần nhất.


````python
i = 1
while True:
    if i > 5:
        break
    print(i)
    i += 1
````

### 2.5. `continue` — bỏ qua lượt này

`continue` quay lại đầu vòng lặp ngay, không chạy phần code phía dưới.


````python
# In các số từ 1 đến 10, BỎ QUA bội của 3
i = 0
while i < 10:
    i += 1
    if i % 3 == 0:
        continue
    print(i)
````

### 2.6. Mẫu thường gặp: nhập đến khi hợp lệ

Một trong những ứng dụng đẹp nhất của `while` là **xác thực đầu vào** (input validation): cứ hỏi đến khi người dùng nhập đúng.


````python
while True:
    s = input('Nhập số nguyên dương: ').strip()
    if s.isdigit() and int(s) > 0:
        n = int(s)
        break
    print('Không hợp lệ, thử lại.')
print(f'Bạn vừa nhập: {n}')
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- `while điều_kiện:` lặp lại khi điều kiện vẫn True.
- Phải có cách thay đổi điều kiện trong thân vòng lặp, kẻo vòng vô hạn.
- `break` thoát vòng lặp gần nhất; `continue` bỏ qua phần còn lại của lượt.
- Mẫu hữu ích: `while True: ... if ...: break` để xác thực đầu vào.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `vòng lặp` | loop | khối lệnh được chạy nhiều lần |
| `while` | while | lặp khi điều kiện True |
| `break` | break | thoát hẳn khỏi vòng lặp |
| `continue` | continue | sang vòng lặp kế tiếp |
| `vòng lặp vô hạn` | infinite loop | lặp mãi vì điều kiện luôn True |
| `biến đếm` | counter | biến tăng/giảm để điều khiển vòng lặp |
| `xác thực đầu vào` | input validation | hỏi đến khi nhập đúng |
