## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Tuple — list 'đông cứng'

**Tuple** giống list nhưng **không đổi được** (immutable). Cú pháp `( )`.

Khi nào dùng tuple?
- Dữ liệu **không nên thay đổi** (toạ độ điểm, ngày tháng năm).
- Trả về **nhiều giá trị** từ một hàm.
- Làm **key cho dict** (list không làm được vì mutable).


````python
diem = (3, 7)
ngay = (22, 5, 2026)
print(diem[0], diem[1])
print(ngay)
# diem[0] = 99   # ❌ TypeError — tuple không đổi được
````

**Packing / Unpacking:** Python cho phép gán nhiều biến cùng lúc rất gọn:


````python
a, b = 1, 2          # packing rồi unpacking
print(a, b)

a, b = b, a          # đổi giá trị 2 biến — không cần biến tạm!
print(a, b)

x, y, z = (10, 20, 30)
print(z)
````

### 2.2. Set — tập hợp phần tử duy nhất

**Set** lưu các phần tử **duy nhất** (không trùng), **không thứ tự**.  
Cú pháp `{1, 2, 3}` (chú ý: `{}` rỗng là dict, set rỗng là `set()`).

Cực kỳ hữu ích để **loại trùng** và làm **phép toán tập hợp**.


````python
s1 = {1, 2, 3, 2, 1}
print(s1)                 # {1, 2, 3} — tự loại trùng

s2 = {3, 4, 5}
print(s1 | s2)            # hợp (union):     {1,2,3,4,5}
print(s1 & s2)            # giao (intersection): {3}
print(s1 - s2)            # hiệu (difference):   {1,2}
print(s1 ^ s2)            # đối xứng:        {1,2,4,5}
````

````python
# Dùng set để loại trùng nhanh
ds = [1, 2, 2, 3, 4, 4, 4, 5]
print(list(set(ds)))
````

### 2.3. Dictionary — ánh xạ key → value

**Dict (từ điển)** lưu cặp **`key: value`**. Truy cập **siêu nhanh** qua key.

Cú pháp `{key1: value1, key2: value2}`.


````python
hoc_sinh = {
    'ten': 'An',
    'tuoi': 16,
    'lop': '10A1',
    'diem': [8, 7.5, 9],
}
print(hoc_sinh['ten'])
print(hoc_sinh['diem'])
hoc_sinh['email'] = 'an@example.com'   # thêm key mới
hoc_sinh['tuoi'] = 17                  # cập nhật value
print(hoc_sinh)
````

### 2.4. Phương thức dict quan trọng

| Method / Cú pháp | Tác dụng |
|------------------|----------|
| `d.keys()` | tất cả key |
| `d.values()` | tất cả value |
| `d.items()` | tất cả cặp (key, value) |
| `d.get(k, default)` | lấy value, mặc định `default` nếu không có |
| `k in d` | kiểm tra có key |
| `del d[k]` | xoá key |
| `d.pop(k, default)` | lấy ra và xoá |


````python
d = {'a': 1, 'b': 2, 'c': 3}
for k, v in d.items():
    print(k, '->', v)

print(d.get('z', 0))     # 0 — không có key 'z'
print('a' in d)          # True
````

### 2.5. Mẫu thường gặp: đếm tần suất

Dùng dict để đếm số lần xuất hiện rất gọn — đây là một mẫu cực kỳ phổ biến.


````python
cau = 'mississippi'
dem = {}
for c in cau:
    dem[c] = dem.get(c, 0) + 1
print(dem)
````

### 2.6. Khi nào dùng cái nào?

| Cấu trúc | Khi nào dùng |
|----------|--------------|
| `list` | nhiều giá trị **có thứ tự**, cho phép trùng, có thể sửa |
| `tuple` | dữ liệu **không nên đổi**, nhóm cố định (toạ độ, ngày...) |
| `set` | chỉ cần biết **có/không**, **duy nhất**, làm phép tập hợp |
| `dict` | ánh xạ **key → value**, tra cứu nhanh |

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- **Tuple** `( )` — không đổi được, dùng cho dữ liệu cố định, packing/unpacking.
- **Set** `{ }` — không trùng, không thứ tự, có phép `| & - ^`. Loại trùng nhanh.
- **Dict** `{k: v}` — ánh xạ key → value, tra cứu nhanh, dùng `.get(k, default)`.
- Mẫu đếm tần suất bằng dict: `dem[x] = dem.get(x, 0) + 1`.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `tuple` | tuple | list không đổi được, ngoặc tròn |
| `tập hợp` | set | phần tử duy nhất, ngoặc nhọn |
| `từ điển` | dict | key → value, ngoặc nhọn |
| `khoá` | key | tên dùng để tra value trong dict |
| `giá trị` | value | dữ liệu được lưu ứng với key |
| `hợp / giao` | union / intersection | phép toán tập hợp |
| `không đổi` | immutable | không sửa được sau khi tạo |
