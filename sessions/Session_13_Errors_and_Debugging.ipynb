### 2.2. Đọc traceback

Python in **traceback** khi gặp lỗi. Cách đọc:

1. **Đọc dòng cuối** — loại lỗi + thông báo.
2. **Đọc các dòng `File ..., line ...`** — nơi lỗi xảy ra.
3. **Đọc từ dưới lên trên** — chuỗi gọi hàm dẫn đến lỗi.

Ví dụ:
```
Traceback (most recent call last):
  File "a.py", line 5, in <module>
    print(divide(10, 0))
  File "a.py", line 2, in divide
    return a / b
ZeroDivisionError: division by zero
```
→ Lỗi `ZeroDivisionError`, xảy ra ở `return a / b` (dòng 2), được gọi từ dòng 5.


### 2.3. `try / except` — bắt lỗi

```python
try:
    # khối có thể lỗi
except LoaiLoi:
    # xử lý khi gặp lỗi
```


````python
try:
    n = int(input('Nhập số: '))
    print(100 / n)
except ValueError:
    print('Bạn nhập không phải số.')
except ZeroDivisionError:
    print('Không chia được cho 0.')
````

### 2.4. Lấy thông tin lỗi: `as e`


````python
try:
    int('abc')
except ValueError as e:
    print('Lỗi:', e)
````

### 2.5. `else` và `finally`

- `else`: chạy nếu **không** có lỗi.
- `finally`: **luôn** chạy, dù có lỗi hay không (đóng tài nguyên...).


````python
try:
    x = int('42')
except ValueError:
    print('Có lỗi')
else:
    print('Không lỗi, x =', x)
finally:
    print('Luôn chạy đoạn này.')
````

### 2.6. Bắt 'mọi' lỗi — cẩn thận!

`except Exception as e:` bắt **gần như mọi lỗi**. Tiện nhưng dễ che bug.  
Quy tắc: **bắt loại lỗi cụ thể** mà bạn biết cách xử lý. Còn lại để Python báo.


````python
# Tệ — che mọi lỗi:
# try:
#     ...
# except:
#     pass

# Tốt — bắt cụ thể:
try:
    s = '12.5'
    print(int(s))
except ValueError:
    print('Không phải số nguyên — thử float?')
    print(float(s))
````

### 2.7. Print debugging — kỹ thuật cổ điển nhưng cực mạnh

Khi không hiểu code chạy ra sao, **rải print** để xem giá trị các biến tại từng bước.


````python
def trung_binh(ds):
    print(f'[debug] ds = {ds}')
    tong = sum(ds)
    print(f'[debug] tong = {tong}')
    n = len(ds)
    print(f'[debug] n = {n}')
    return tong / n

print(trung_binh([8, 7, 9]))
````

### 2.8. Giả thuyết khi debug / Hypothesis-driven debugging

1. **Tái hiện** lỗi (chạy lại với input gây lỗi).
2. **Đặt giả thuyết** về nguyên nhân (vd: 'tôi nghĩ biến x = None').
3. **Kiểm tra** giả thuyết bằng `print` hoặc thử nghiệm.
4. **Sửa** rồi xác nhận lỗi đã hết.
5. **Viết test** để bug không quay lại.
