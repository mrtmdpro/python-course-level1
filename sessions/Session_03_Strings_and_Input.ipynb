## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Đọc dữ liệu từ người dùng / `input()`

`input()` dừng chương trình, chờ người dùng gõ một dòng, rồi trả về **chuỗi (string)** đó.

Lưu ý quan trọng: `input()` **luôn trả về chuỗi**. Muốn dùng làm số, phải đổi kiểu.


````python
ten = input('Bạn tên gì? ')
print('Xin chào,', ten + '!')
````

````python
# Nhập tuổi và tính tuổi sang năm sau
tuoi = int(input('Bạn bao nhiêu tuổi? '))
print('Năm sau bạn sẽ', tuoi + 1, 'tuổi.')
````

### 2.2. Index — truy cập từng ký tự

Trong Python, **index bắt đầu từ 0** (không phải 1!).

```
  P  y  t  h  o  n
  0  1  2  3  4  5
 -6 -5 -4 -3 -2 -1   <- index âm: đếm ngược từ cuối
```


````python
s = 'Python'
print(s[0])     # 'P'
print(s[3])     # 'h'
print(s[-1])    # 'n'  — ký tự cuối
print(s[-2])    # 'o'
print(len(s))   # 6 — số ký tự
````

### 2.3. Slicing — cắt chuỗi

Cú pháp: `s[start:end:step]` — lấy từ `start` đến **trước** `end`, mỗi `step` ký tự.

Bỏ qua mặc định: `start=0`, `end=len(s)`, `step=1`.


````python
s = 'Lập trình Python'
print(s[0:3])      # 'Lập'
print(s[4:9])      # 'trình'
print(s[:3])       # 'Lập' — từ đầu
print(s[10:])      # 'Python' — đến cuối
print(s[::2])      # bước 2 ký tự
print(s[::-1])     # đảo ngược chuỗi
````

### 2.4. Phương thức chuỗi / String methods

**Phương thức (method)** là một hàm gắn liền với một đối tượng. Cú pháp: `chuỗi.tên_phương_thức(...)`.

Một số phương thức hữu ích:

| Method | Tác dụng |
|--------|----------|
| `s.upper()` | viết HOA toàn bộ |
| `s.lower()` | viết thường toàn bộ |
| `s.strip()` | xoá khoảng trắng đầu / cuối |
| `s.split(sep)` | tách chuỗi thành **list** theo `sep` |
| `sep.join(list)` | nối list lại thành chuỗi |
| `s.replace(a, b)` | thay `a` bằng `b` |
| `s.find(x)` | tìm vị trí đầu của `x` (`-1` nếu không có) |
| `s.count(x)` | đếm số lần `x` xuất hiện |
| `s.startswith(x)` / `s.endswith(x)` | bắt đầu / kết thúc bằng `x`? |


````python
s = '   Học Python rất vui!   '
print(s.strip())
print(s.strip().upper())
print(s.replace('vui', 'hay'))
print(s.strip().split(' '))
print(s.count('o'))
````

### 2.5. f-string — định dạng chuỗi đẹp

Đặt chữ `f` trước dấu `'`, rồi nhúng biến hoặc biểu thức trong `{...}`.


````python
ten = 'Lan'
tuoi = 16
diem = 8.7
print(f'Bạn {ten}, {tuoi} tuổi, đạt điểm {diem}.')
print(f'Năm sau bạn {tuoi + 1} tuổi.')
print(f'Điểm làm tròn 1 chữ số: {diem:.1f}')   # định dạng số
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- `input(prompt)` đọc dòng người dùng — luôn trả về **chuỗi**.
- Index bắt đầu từ 0; index âm đếm ngược từ cuối.
- Slice `s[a:b:step]` cắt chuỗi (không gồm `b`).

- f-string `f'... {bien} ...'` định dạng chuỗi gọn và rõ.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |

| `chỉ số` | index | vị trí của ký tự, bắt đầu từ 0 |
| `cắt chuỗi` | slicing | cú pháp `s[a:b:step]` |

| `chuỗi định dạng` | f-string | đặt `f` trước, nhúng biến trong `{}` |
| `nối chuỗi` | concatenation | dùng `+` để gộp 2 chuỗi |
| `độ dài` | len | `len(s)` trả về số ký tự |
