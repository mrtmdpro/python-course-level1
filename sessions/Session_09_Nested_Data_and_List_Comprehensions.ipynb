## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. List 2D — list of lists

Một list có thể chứa list khác → thành **bảng 2 chiều**.


````python
diem = [
    [8, 7, 9],     # bạn 0
    [6, 5, 7],     # bạn 1
    [10, 9, 8],    # bạn 2
]
print(diem[0])         # [8, 7, 9]
print(diem[0][2])      # 9 — bạn 0, môn 2
print(len(diem))       # 3 hàng
print(len(diem[0]))    # 3 cột
````

### 2.2. Duyệt list 2D

Dùng vòng `for` lồng — vòng ngoài duyệt **hàng**, vòng trong duyệt **cột**.


````python
diem = [[8, 7, 9], [6, 5, 7], [10, 9, 8]]
for hang in diem:
    for x in hang:
        print(x, end=' ')
    print()

# In trung bình mỗi học sinh
for i, hang in enumerate(diem):
    tb = sum(hang) / len(hang)
    print(f'HS {i}: TB = {tb:.2f}')
````

### 2.3. List of dicts — dữ liệu kiểu bảng

Cách phổ biến nhất để biểu diễn 1 bảng dữ liệu trong Python: list các dict, mỗi dict là 1 dòng.


````python
hoc_sinh = [
    {'ten': 'An', 'lop': '10A1', 'diem': 8.5},
    {'ten': 'Bình', 'lop': '10A2', 'diem': 7.0},
    {'ten': 'Châu', 'lop': '10A1', 'diem': 9.0},
    {'ten': 'Dương', 'lop': '10A3', 'diem': 6.5},
]
for h in hoc_sinh:
    print(f"{h['ten']:10s} {h['lop']:6s} {h['diem']}")
````

### 2.4. Lọc và sắp xếp

Dùng `sorted(ds, key=...)` để sắp theo một trường, dùng vòng `for + if` để lọc.


````python
hoc_sinh = [
    {'ten': 'An', 'diem': 8.5},
    {'ten': 'Bình', 'diem': 7.0},
    {'ten': 'Châu', 'diem': 9.0},
    {'ten': 'Dương', 'diem': 6.5},
]

# Sắp theo điểm giảm dần
theo_diem = sorted(hoc_sinh, key=lambda h: h['diem'], reverse=True)
for h in theo_diem:
    print(h['ten'], h['diem'])
````

### 2.5. List comprehension — siêu vũ khí

Cú pháp gọn để **tạo list mới** từ list cũ.

```python
[biểu_thức for x in dữ_liệu if điều_kiện]
```

So sánh viết kiểu cũ và kiểu comprehension:


````python
# Cách cũ — 4 dòng
binh_phuong = []
for x in range(1, 6):
    binh_phuong.append(x * x)
print(binh_phuong)

# Comprehension — 1 dòng
binh_phuong = [x * x for x in range(1, 6)]
print(binh_phuong)

# Có điều kiện
chan = [x for x in range(20) if x % 2 == 0]
print(chan)

# Biến đổi từng phần tử
ten = ['  An  ', 'bình', 'CHÂU']
chuan = [t.strip().title() for t in ten]
print(chuan)
````

### 2.6. Dict comprehension và set comprehension (xem nhanh)


````python
# Dict comp
binh = {x: x * x for x in range(1, 6)}
print(binh)

# Set comp
ng = {c for c in 'Học Python rất vui' if c in 'aeiou'}
print(ng)
````

## 8. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- Dữ liệu lồng: list của list, list của dict.
- Vòng for lồng để duyệt 2D: ngoài là hàng, trong là cột.
- List comprehension: `[expr for x in data if cond]` — gọn và mạnh.
- Cũng có dict / set comprehension. Tránh comprehension quá phức tạp.
- `sorted(data, key=lambda h: h['field'])` để sắp theo trường.

### Từ vựng / Vocabulary

| `lồng nhau` | nested | list trong list, dict trong dict, ... |
| `danh sách 2 chiều` | 2D list | list các list cùng độ dài |
| `list comprehension` | list comprehension | cú pháp gọn để tạo list |
| `lambda` | lambda | hàm 1 dòng, dùng cho key= của sort/min/max |
