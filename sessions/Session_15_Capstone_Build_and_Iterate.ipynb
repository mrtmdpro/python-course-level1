## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Hoạt động: Code review chéo (15 phút)

Theo cặp:
1. Đổi notebook cho nhau (chia sẻ Colab link / màn hình).
2. Trong 5 phút, **đọc** code của bạn → ghi 3 điểm:
   - 1 điểm **tốt** (clean, đẹp, rõ ràng)
   - 1 điểm **có thể cải thiện**
   - 1 **câu hỏi** tò mò
3. Đổi vai trong 5 phút.
4. 5 phút thảo luận: cùng nhau xem 1-2 chỗ có thể sửa ngay.

**Quy tắc review tử tế:**
- Khen trước, góp ý sau.
- Nói về **code**, không phải **người**.
- Đề xuất, không ra lệnh: 'Có thể thử...' thay vì 'Phải sửa...'


### 2.2. Edge cases — các trường hợp biên

Hãy tự hỏi cho dự án của bạn:

**Track A (Quiz):**
- Người chơi nhập rỗng / khoảng trắng.
- Người chơi nhập sai dạng (chọn `E` khi chỉ có A-D).
- Có 0 câu hỏi (list rỗng).
- Tên có dấu cách thừa hai đầu.

**Track B (Data):**
- File không tồn tại.
- File có ô trống / NaN.
- Cột không phải số (string lẫn vào).
- File chỉ có 1 dòng.

**Track C (Turtle):**
- Người dùng nhập kích thước âm hoặc 0.
- Số đa giác < 3 cạnh.
- Cửa sổ bị đóng giữa chừng.


### 2.3. Refactor checklist

Trong 30 phút giữa buổi, hãy đi qua checklist:

- [ ] Mỗi hàm làm **một việc** rõ ràng?
- [ ] Tên hàm và biến **có ý nghĩa**?
- [ ] Có **docstring** ngắn cho mỗi hàm?
- [ ] Code có thụt lề **nhất quán** (4 spaces)?
- [ ] Đã tách **dữ liệu** khỏi **logic**? (ví dụ: list QUESTIONS để riêng, không lẫn trong logic)
- [ ] Đã có ít nhất 1 chỗ dùng `try/except` cho input?
- [ ] Comment giải thích **tại sao**, không phải **làm gì** (làm gì thì code đã nói)?


### 2.4. Mẹo nhỏ theo track

**Track A — Quiz:**
- Tách `QUESTIONS` ra file `data.py` hoặc `questions.csv` để dễ thêm câu hỏi.
- Dùng `random.sample(QUESTIONS, 5)` để chọn 5 câu ngẫu nhiên (không trùng).
- Xếp loại bằng dict tra cứu: `{0.9: 'Excellent', 0.7: 'Good', ...}`.

**Track B — Data:**
- `df.dropna()` xoá các dòng có ô trống.
- `df['col'].astype(float)` ép kiểu khi cần.
- Đặt `plt.tight_layout()` để biểu đồ không bị cắt nhãn.
- Lưu biểu đồ: `plt.savefig('bieu_do.png', dpi=150, bbox_inches='tight')`.

**Track C — Turtle:**
- `t.speed(0)` — tốc độ tối đa.
- `screen.tracer(0)` — tắt animation, vẽ tức thì; cuối cùng `screen.update()`.
- Đóng gói các hình thành **hàm** (`ve_sao`, `ve_hoa_thi`, ...) — dễ tổ hợp.


### 2.5. Test thủ công

Liệt kê **3 kịch bản test** cho dự án và chạy thử:

Ví dụ Track A:
1. **Happy path**: nhập đúng 100% câu → kỳ vọng 'Excellent'.
2. **Sai 100%**: nhập sai hết → kỳ vọng 'Try again'.
3. **Input rỗng**: bấm Enter → không crash, hỏi lại.

Sau mỗi kịch bản, ghi: **kỳ vọng** vs **thực tế**.

## 7. Tổng kết & Từ vựng (Recap & Glossary)

### Bạn đã học được gì? / What you learned
- Code review chéo giúp nhìn ra điểm mù trong code mình.
- Edge cases là chìa khoá phân biệt code 'demo' và code 'có thể tin được'.
- Refactor: hàm ngắn, tên rõ, docstring — code đẹp tự nhiên.

### Từ vựng / Vocabulary

| English | Tiếng Việt | Ghi chú / Note |
|---------|------------|----------------|
| `review code` | code review | đọc và góp ý code của người khác |
| `trường hợp biên` | edge case | input bất thường mà code phải chịu |
| `tái cấu trúc` | refactor | sửa lại code cho gọn, không đổi hành vi |
| `docstring` | docstring | chuỗi mô tả ngay sau def |
| `MVP → polish` | MVP → polish | đi từ chạy được đến đẹp & vững |
