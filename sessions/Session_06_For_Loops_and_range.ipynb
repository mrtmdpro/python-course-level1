## 2. Khái niệm chính (Key concepts) + Live coding

### 2.1. Vòng lặp `for`

`for` dùng để **duyệt qua** từng phần tử của một tập hợp (chuỗi, list, range...).

Cú pháp:

```python
for biến in tập_hợp:
    # khối lệnh
```


````python
for ky_tu in 'Python':
    print(ky_tu)
````

````python
for mon in ('Toán', 'Lý', 'Hoá'):
    print(f'Hôm nay học {mon}')
````

### 2.2. `range()` — sinh dãy số

`range()` cho ra một dãy số nguyên.  
Quan trọng: số `stop` **KHÔNG** được tính.

| Cú pháp | Sinh ra |
|---------|---------|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 7)` | 2, 3, 4, 5, 6 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -1)` | 10, 9, 8, ..., 1 |


````python
for i in range(5):
    print(i)
````

````python
# In bảng cửu chương 7
for i in range(1, 11):
    print(f'7 x {i} = {7 * i}')
````

### 2.3. Mẫu accumulator (tích / tổng / đếm)

**Mẫu chung** xuất hiện rất nhiều: tạo 1 biến tích luỹ, rồi cập nhật trong mỗi lượt lặp.


````python
# Tổng các số từ 1 đến 100
tong = 0
for i in range(1, 101):
    tong += i
print('Tổng =', tong)
````

````python
# Đếm số chữ 'o' trong câu
s = 'Học Python rất vui và bổ ích.'
dem = 0
for c in s:
    if c == 'o':
        dem += 1
print(dem)
````

### 2.4. Vòng lặp lồng nhau / Nested loops

Một vòng lặp có thể nằm bên trong một vòng lặp khác. Tổng số lần chạy = tích các vòng.


````python
# Bảng cửu chương 2 → 5
for a in range(2, 6):
    for b in range(1, 11):
        print(f'{a} x {b} = {a * b}')
    print('-' * 20)
````

````python
# Vẽ tam giác sao
n = 5
for i in range(1, n + 1):
    print('*' * i)
````

### 2.5. `for ... else` (tuỳ chọn)

Khối `else` của `for` chỉ chạy nếu vòng lặp **kết thúc tự nhiên** (không qua `break`).


````python
# Tìm chữ 'z' trong câu
s = 'Học Python rất vui'
for c in s:
    if c == 'z':
        print('Có z!')
        break
else:
    print('Không có z.')
````
